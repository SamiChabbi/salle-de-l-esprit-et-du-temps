## 4. Le schéma à connaître par cœur

**Diviser pour régner**, c'est trois étapes :
1. **Diviser** le problème en sous-problèmes plus petits et indépendants (le plus souvent : couper la liste en deux moitiés) ;
2. **Régner** : résoudre chaque sous-problème, par un **appel récursif** (jusqu'à un cas de base assez petit pour être résolu directement) ;
3. **Combiner** les solutions des sous-problèmes pour obtenir la solution du problème de départ.

**Le gabarit**, qui sert pour presque toutes les questions D4 de la base :
```python
def resoudre(liste):
    if len(liste) <= 1:                             # cas de base : réponse directe
        return ...
    gauche = resoudre(moitie_gauche(liste))         # diviser + régner
    droite = resoudre(moitie_droite(liste))
    return combiner(gauche, droite)                 # combiner
```
**La même chose avec des indices** (sans fabriquer de nouvelles listes), comme dans le tri fusion de Polynésie 2021, Stooge ou le labyrinthe :
```python
def resoudre(liste, debut, fin):
    if debut >= fin:                                # au plus un élément
        return ...
    milieu = (debut + fin) // 2
    gauche = resoudre(liste, debut, milieu)
    droite = resoudre(liste, milieu + 1, fin)
    return combiner(...)
```
**Et la dichotomie, qui ne garde qu'une moitié** : on regarde le milieu, et on ne fait **qu'un seul** appel (ou un seul tour de boucle) sur la moitié où peut se trouver la réponse.

### Fiche D4 — Écrire ou compléter la fonction qui divise et s'appelle elle-même

🔎 **Je la reconnais :** « Écrire la fonction récursive… », « On procédera de la façon suivante : tester si le cas de base est atteint ; découper la liste en deux… ; appeler récursivement… ; comparer les deux valeurs… », « Recopier et compléter les lignes 8, 9 et 10 », « Finaliser le programme en complétant les lignes 6, 7 et 8 ».

🛠️ **La méthode :**
1. **L'énoncé donne presque toujours les étapes en français** (une liste à tirets). On les numérote, et on écrit **une ligne de code par tiret**, dans l'ordre.
2. Sinon, on remplit le gabarit : quel est le **cas de base** ? comment **couper** ? comment **combiner** les deux résultats ?
3. On vérifie sur une liste de 2 ou 3 éléments, avec l'arbre des appels.

📌 **Exemple : le maximum d'une liste par diviser pour régner** (même structure que `plus_court_segment` de l'exercice 14).

- Cas de base : une liste d'un seul élément → c'est le maximum.
- Diviser : moitié gauche, moitié droite.
- Combiner : le plus grand des deux maximums.

````python
def moitie_gauche(t):
    res = []
    for i in range(len(t) // 2):
        res.append(t[i])
    return res


def moitie_droite(t):
    res = []
    for i in range(len(t) // 2, len(t)):
        res.append(t[i])
    return res


def maximum(t):
    if len(t) == 1:                       # cas de base
        return t[0]
    m_g = maximum(moitie_gauche(t))       # diviser + régner
    m_d = maximum(moitie_droite(t))
    if m_g >= m_d:                        # combiner
        return m_g
    else:
        return m_d


print(maximum([3, 8, 2, 9, 4, 1]))
````

⚠️ **Pièges relevés dans la base :**
- **le cas de base qui ne coupe plus** : si une moitié peut avoir la même taille que la liste de départ (une liste d'un élément dont la « moitié gauche » est elle-même), la récursion ne s'arrête jamais → le cas de base doit arriver **avant** ;
- avec des indices, la partie droite commence à `milieu + 1`, **pas** à `milieu` ;
- en combinant, un élément peut être compté **deux fois** (Douglas-Peucker : le point sélectionné est au bout des deux morceaux) ;
- renvoyer l'**objet** demandé (le segment) et pas seulement sa valeur (sa longueur).

🛠️ **La méthode :**
- **Pour un tri fusion ou tout algorithme à 2 appels : l'arbre des appels.** On écrit la liste en haut, ses deux moitiés en dessous, etc., jusqu'aux listes d'un élément. Puis, si on demande les **fusions**, on remonte en fusionnant.
- **L'ordre des affichages** = l'ordre où on visite l'arbre : la racine, puis **toute** la branche gauche, puis la branche droite (pas étage par étage).
- **Pour une dichotomie : un tableau** avec une ligne par appel (ou par tour) et les colonnes `debut`, `fin`, `milieu`, la valeur au milieu, la décision.

📌 **Exemple.** Les appels qui affichent dans `tri_fusion([5, 2, 8, 1, 9, 3])`, quand la fonction fait `print(L)` au début de chaque appel qui n'est pas un cas de base :
```
                 [5, 2, 8, 1, 9, 3]
                 /                \
          [5, 2, 8]            [1, 9, 3]
          /       \            /       \
        [5]     [2, 8]       [1]     [9, 3]
```
Affichages, dans l'ordre : `[5, 2, 8, 1, 9, 3]`, `[5, 2, 8]`, `[2, 8]`, `[1, 9, 3]`, `[9, 3]`. (Les listes d'un seul élément sont des cas de base : elles n'affichent rien.)

### Fiche D5 — Couper en deux : moitiés, milieu, indices

🔎 **Je la reconnais :** « Écrire la fonction `moitie_gauche` / `moitie_droite` », « Déterminer la valeur prise par `k` », « Si le nombre d'éléments est impair, l'élément du centre se trouve dans la partie gauche », une recherche avec `debut`, `fin`, `milieu`.

🛠️ **Ce qu'il faut savoir écrire sans réfléchir :**

| Besoin | Code |
|:---|:---|
| moitié gauche (centre à droite) | `for i in range(len(t) // 2): res.append(t[i])` |
| moitié droite (centre à droite) | `for i in range(len(t) // 2, len(t)): res.append(t[i])` |
| moitié gauche (centre à gauche) | `for i in range((len(t) + 1) // 2): res.append(t[i])` |
| le milieu entre deux indices | `milieu = (debut + fin) // 2` |
| on garde la droite / la gauche | `debut = milieu + 1` / `fin = milieu - 1` |
| copie de `i` à `j` inclus | `for k in range(i, j + 1): res.append(t[k])` |

⚠️ **Pièges :** `//` (division entière) et pas `/` ; oublier le `+ 1` ou le `- 1` dans la dichotomie (la zone ne rétrécit plus : boucle infinie) ; `while debut <= fin` avec `<=`.

### Fiche D3 — La fusion (l'étape « combiner » du tri fusion)

🔎 **Je la reconnais :** « Écrire la fonction `fusion` qui prend deux listes triées… », « Écrire les instructions manquantes des lignes 17 à 22 », « Parmi les propositions, quelle instruction placer à la ligne 4 ? », « Expliquer comment rassembler deux tas de cartes triés ».

🛠️ **Le modèle des deux indices** : `i1` avance dans la première liste, `i2` dans la seconde ; à chaque tour on compare `L1[i1]` et `L2[i2]`, on ajoute **le plus petit**, et on avance **seulement** son indice. Quand une liste est finie, on recopie **le reste** de l'autre.

Il existe **deux façons de l'écrire**, et les deux tombent au bac :

| Version « `and` + reste » | Version « `or` + 3 cas » |
|:---|:---|
| `while i1 < len(L1) and i2 < len(L2):` comparer, ajouter le plus petit | `while i1 < len(L1) or i2 < len(L2):` |
| puis deux boucles pour recopier le reste de `L1`, puis de `L2` | `if i1 >= len(L1):` prendre dans `L2` / `elif i2 >= len(L2):` prendre dans `L1` / `else:` comparer |

⚠️ **Pièges :** oublier le reste (testé par `fusion([1, 2, 3], [])`) ; avancer les deux indices en même temps ; mélanger `and` et `or` (avec `and` **et** sans les boucles du reste, on perd des éléments).

### Fiche D1 — Reconnaître la méthode

🔎 **Je la reconnais :** « En informatique, le principe diviser pour régner signifie… », « Indiquer le qualificatif qui s'applique le mieux à l'algorithme », « Cet algorithme est-il itératif ou récursif ? Justifier ».

| Méthode | Ce qu'on voit |
|:---|:---|
| **diviser pour régner** | on coupe en morceaux plus petits et indépendants, on traite **tous** les morceaux, puis on **combine** |
| **dichotomique** | on coupe en deux et on ne garde **qu'une** moitié |
| **glouton** | à chaque étape, on prend le choix qui semble le meilleur **tout de suite**, sans revenir en arrière |
| **récursif** | l'algorithme s'utilise lui-même (« trier les deux parties avec le tri fusion ») |
| **itératif** | des boucles, pas d'appel à soi-même |

⚠️ **Piège :** « diviser une fonction en deux fonctions plus petites » n'est **pas** diviser pour régner : on découpe le **problème**, pas le programme.

## 6. Et l'exponentiation rapide ?

Elle est souvent donnée en cours comme exemple de diviser pour régner, mais **elle n'est tombée dans aucun des sujets de la base**. Si elle est dans ton cours, elle se traite avec la même méthode :
- cas de base : x⁰ = 1 ;
- diviser : n // 2 ;
- combiner : si n est pair, xⁿ = x^(n/2) × x^(n/2) ; s'il est impair, xⁿ = x × x^(n//2) × x^(n//2).

On calcule `x^(n//2)` **une seule fois** (dans une variable) : c'est ce qui rend l'algorithme rapide (log₂ n multiplications au lieu de n).

````python
def puissance_rapide(x, n):
    if n == 0:
        return 1
    moitie = puissance_rapide(x, n // 2)     # un seul appel récursif
    if n % 2 == 0:
        return moitie * moitie
    else:
        return x * moitie * moitie


print(puissance_rapide(2, 10), puissance_rapide(3, 5))
````

## Exercice 2 — Fusionner deux listes triées  ⭐

📚 **Tombé au bac :** épreuve pratique **2025, sujet 12**, exercice 1 (et 2024, sujet 34 ; 2023, sujet 24 ; 2022, sujet 30)

### Question 1

🏷️ *Type de question :* **D3** Écrire ou compléter la fusion (l'étape « combiner »)

#### 📝 ÉNONCÉ

> Programmer la fonction `fusion` prenant en paramètres deux tableaux non vides `tab1` et `tab2` (type `list`) d'entiers, chacun dans l'ordre croissant, et renvoyant un tableau trié dans l'ordre croissant et contenant l'ensemble des valeurs de `tab1` et `tab2`.
>
> Exemples :
> ```python
> >>> fusion([3, 5], [2, 5])
> [2, 3, 5, 5]
> >>> fusion([-2, 4], [-3, 5, 10])
> [-3, -2, 4, 5, 10]
> >>> fusion([4], [2, 6])
> [2, 4, 6]
> >>> fusion([], [])
> []
> >>> fusion([1, 2, 3], [])
> [1, 2, 3]
> ```

#### ✅ RÉSOLUTION

````python
def fusion(tab1, tab2):
    tab_fusion = []
    i1 = 0
    i2 = 0
    while i1 < len(tab1) and i2 < len(tab2):
        if tab1[i1] < tab2[i2]:
            tab_fusion.append(tab1[i1])
            i1 = i1 + 1
        else:
            tab_fusion.append(tab2[i2])
            i2 = i2 + 1
    while i1 < len(tab1):          # il reste des éléments dans tab1
        tab_fusion.append(tab1[i1])
        i1 = i1 + 1
    while i2 < len(tab2):          # il reste des éléments dans tab2
        tab_fusion.append(tab2[i2])
        i2 = i2 + 1
    return tab_fusion
````

*Les tests de la feuille d'exercices :*

````python
assert fusion([3, 5], [2, 5]) == [2, 3, 5, 5]
assert fusion([-2, 4], [-3, 5, 10]) == [-3, -2, 4, 5, 10]
assert fusion([4], [2, 6]) == [2, 4, 6]
assert fusion([], []) == []
assert fusion([1, 2, 3], []) == [1, 2, 3]
assert fusion([], [7, 8]) == [7, 8]
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Le modèle des deux indices** : `i1` avance dans `tab1`, `i2` dans `tab2`. À chaque tour, on compare `tab1[i1]` et `tab2[i2]`, on recopie le **plus petit**, et on avance **seulement** l'indice d'où il vient.
- Quand une des deux listes est finie, on recopie **le reste** de l'autre (les deux dernières boucles ; une seule des deux fait quelque chose).
- Tableau de trace pour `fusion([3, 5], [2, 5])` :

| tour | `i1` | `i2` | comparaison | `tab_fusion` |
|:---|:---|:---|:---|:---|
| départ | 0 | 0 | | `[]` |
| 1 | 0 | 1 | 3 < 2 ? non → on prend 2 | `[2]` |
| 2 | 1 | 1 | 3 < 5 ? oui → on prend 3 | `[2, 3]` |
| 3 | 1 | 2 | 5 < 5 ? non → on prend 5 (de `tab2`) | `[2, 3, 5]` |
| reste | 2 | 2 | `tab2` fini → on recopie 5 | `[2, 3, 5, 5]` |

- ⚠️ Oublier de recopier le reste est l'erreur la plus fréquente : le test `fusion([1, 2, 3], [])` la détecte.
- C'est la fonction **combiner** du tri fusion : elle revient dans presque tous les exercices de ce thème.

## Exercice 3 — Recherche dichotomique  ⭐

📚 **Tombé au bac :** épreuve pratique **2025, sujet 10**, exercice 1 (la même recherche, à trous, est aussi tombée en 2025 aux sujets 9 et 18)

### Question 1

🏷️ *Type de question :* **D4** Écrire ou compléter la fonction qui divise et s'appelle elle-même · **D5** Couper en deux : moitiés, milieu, indices

#### 📝 ÉNONCÉ

> Écrire une fonction `recherche` qui prend en paramètres un tableau `tab` de nombres entiers triés par ordre croissant et un nombre entier `n`, et qui effectue une recherche dichotomique du nombre entier `n` dans le tableau non vide `tab`.
>
> Cette fonction doit renvoyer un indice correspondant au nombre cherché s'il est dans le tableau, `None` sinon.
>
> Exemples :
> ```python
> >>> recherche([2, 3, 4, 5, 6], 5)
> 3
> >>> recherche([2, 3, 4, 6, 7], 5) # renvoie None
> ```

#### ✅ RÉSOLUTION

````python
def recherche(tab, n):
    ind_debut = 0
    ind_fin = len(tab) - 1
    while ind_debut <= ind_fin:
        ind_milieu = (ind_debut + ind_fin) // 2
        if tab[ind_milieu] == n:
            return ind_milieu
        elif tab[ind_milieu] < n:
            ind_debut = ind_milieu + 1
        else:
            ind_fin = ind_milieu - 1
    return None
````

*Les tests de la feuille d'exercices :*

````python
assert recherche([2, 3, 4, 5, 6], 5) == 3
assert recherche([2, 3, 4, 6, 7], 5) is None
assert recherche([2, 3, 4, 5, 6], 2) == 0
assert recherche([2, 3, 4, 5, 6], 6) == 4
assert recherche([8], 8) == 0
assert recherche([8], 3) is None
grand = []
for k in range(0, 1000000, 3):
    grand.append(k)
assert recherche(grand, 999999) == 333333
assert recherche(grand, 1000) is None
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **La zone de recherche** est entre `ind_debut` et `ind_fin` (inclus). On regarde le milieu : trouvé → on renvoie l'indice ; trop petit → on garde la moitié **droite** ; trop grand → la moitié **gauche**.
- Tableau de trace pour `recherche([2, 3, 4, 5, 6], 5)` :

| tour | `ind_debut` | `ind_fin` | `ind_milieu` | `tab[ind_milieu]` | décision |
|:---|:---|:---|:---|:---|:---|
| 1 | 0 | 4 | 2 | 4 | 4 < 5 → `ind_debut = 3` |
| 2 | 3 | 4 | 3 | 5 | trouvé → renvoie 3 |

- ⚠️ Trois pièges : `while ind_debut <= ind_fin` (avec `<=`, sinon on rate la dernière case) ; `ind_milieu + 1` et `ind_milieu - 1` (sans le ±1, la zone peut ne plus rétrécir : boucle infinie) ; `return None` **après** la boucle, pas dedans.
- Coût : à chaque tour la zone est divisée par 2 → environ log₂(n) tours. Pour un million d'éléments : 20 tours au lieu d'un million.
- C'est de la **dichotomie** : on coupe en deux et on ne garde qu'**une** moitié. Le programme la range dans « diviser pour régner ».

## Exercice 4 — Recherche dichotomique récursive  ⭐⭐

📚 **Tombé au bac :** épreuve pratique **2024, sujet 18**, exercice 2 (et 2025, sujet 15)

### Question

🏷️ *Type de question :* **D4** Écrire ou compléter la fonction qui divise et s'appelle elle-même

#### 📝 ÉNONCÉ

> Soit `tab` un tableau non vide d'entiers triés dans l'ordre croissant et `n` un entier.
>
> La fonction `chercher` ci-dessous doit renvoyer un indice où la valeur `n` apparaît dans `tab` si cette valeur y figure et `None` sinon.
>
> Les paramètres de la fonction sont :
> - `tab`, le tableau dans lequel s'effectue la recherche ;
> - `x`, l'entier à chercher dans le tableau ;
> - `i`, l'indice de début de la partie du tableau où s'effectue la recherche ;
> - `j`, l'indice de fin de la partie du tableau où s'effectue la recherche.
>
> L'algorithme demandé est une recherche dichotomique récursive.
>
> Recopier et compléter le code de la fonction `chercher` suivante :
> ```python
> 1  def chercher(tab, x, i, j):
> 2      '''Renvoie l'indice de x dans tab, si x est dans tab,
> 3      None sinon.
> 4      On suppose que tab est trié dans l'ordre croissant.'''
> 5      if i > j:
> 6          return None
> 7      m = (i + j) // ...
> 8      if ... < x:
> 9          return chercher(tab, x, ... , ...)
> 10     elif tab[m] > x:
> 11         return chercher(tab, x, ... , ...)
> 12     else:
> 13         return ...
> ```
> Exemples :
> ```python
> >>> chercher([1, 5, 6, 6, 9, 12], 7, 0, 10)
> >>> chercher([1, 5, 6, 6, 9, 12], 7, 0, 5)
> >>> chercher([1, 5, 6, 6, 9, 12], 9, 0, 5)
> 4
> >>> chercher([1, 5, 6, 6, 9, 12], 6, 0, 5)
> 2
> ```

#### ✅ RÉSOLUTION

````python
def chercher(tab, x, i, j):
    '''Renvoie l'indice de x dans tab, si x est dans tab,
    None sinon.
    On suppose que tab est trié dans l'ordre croissant.'''
    if i > j:
        return None
    m = (i + j) // 2
    if tab[m] < x:
        return chercher(tab, x, m+1 , j)
    elif tab[m] > x:
        return chercher(tab, x, i , m-1)
    else:
        return m
````

*Les tests de la feuille d'exercices :*

````python
assert chercher([1, 5, 6, 6, 9, 12], 7, 0, 10) is None
assert chercher([1, 5, 6, 6, 9, 12], 7, 0, 5) is None
assert chercher([1, 5, 6, 6, 9, 12], 9, 0, 5) == 4
assert chercher([1, 5, 6, 6, 9, 12], 6, 0, 5) == 2
assert chercher([1, 5, 6, 6, 9, 12], 1, 0, 5) == 0
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Le milieu `m` ; si `tab[m]` est trop petit, on cherche à droite (`m+1` à `j`) ; trop grand, à gauche (`i` à `m-1`) ; sinon on a trouvé : on renvoie l'indice `m`.
- Le cas de base `i > j` (zone vide) est déjà écrit : c'est lui qui renvoie `None`.
- Tableau de trace pour `chercher([1, 5, 6, 6, 9, 12], 9, 0, 5)` : (i=0, j=5, m=2, 6 < 9) → (i=3, j=5, m=4, 9 = 9) → renvoie 4.
- Le premier exemple a `j = 10`, plus grand que le dernier indice : ça marche quand même ici, car on ne lit jamais `tab[m]` avec `m` trop grand.

Nous rappelons qu'en langage Python la comparaison de chaînes de caractères est possible à l'aide des opérateurs classiques de comparaison : `<`, `>`, `==`, `>=` et `<=`. Ces opérateurs utilisent l'ordre alphabétique.

Par exemple, l'expression booléenne `'ANNEE' < 'BATEAU'` vaut `True` puisque le mot 'ANNEE' est placé avant le mot 'BATEAU' dans l'ordre alphabétique.

Voici le code d'une fonction `ajouter_mot_liste(liste, mot)` qui a pour paramètres `liste`, une liste de chaînes de caractères (ce sont les mots) et `mot`, une chaîne de caractères correspondant au mot que l'on souhaite ajouter à `liste`. Cette fonction modifie `liste` en y ajoutant `mot` selon l'ordre alphabétique. Elle renvoie de plus `liste` après cet ajout.
```python
1 def ajouter_mot_liste(liste, mot):
2     i = 0
3     while i < len(liste):
4         if mot < liste[i]:
5             # La méthode "insert" permet d'ajouter un
6             # élément à un indice donné dans "liste",
7             # les éléments suivants sont décalés.
8             liste.insert(i, mot)
9             return liste
10        i = i + 1
11    # La méthode "append" permet d'ajouter un nouvel
12    # élément à la fin de "liste".
13    liste.append(mot)
14    return liste
```
Voici des exemples d'utilisation de cette fonction :
```python
>>> ajouter_mot_liste([],'NSI')
['NSI']
>>> ajouter_mot_liste(['NSI'],'PYTHON')
['NSI','PYTHON']
>>> ajouter_mot_liste(['NSI','PYTHON'],'OBJET')
['NSI','OBJET','PYTHON']

```

*La fonction `ajouter_mot_liste` du sujet (à exécuter) :*

````python
def ajouter_mot_liste(liste, mot):
    i = 0
    while i < len(liste):
        if mot < liste[i]:
            liste.insert(i, mot)
            return liste
        i = i + 1
    liste.append(mot)
    return liste
````

**Partie C**

Nous allons maintenant voir comment faire la recherche d'un mot dans notre dictionnaire.

Nous souhaitons une recherche rapide et nous allons profiter du fait que les mots sont classés dans l'ordre alphabétique dans les listes du dictionnaire.

Nous prendrons une approche dichotomique pour écrire une fonction `est_present(liste, mot, debut, fin)` qui a pour paramètres :
- `liste`, une liste de mots classés dans l'ordre alphabétique ;
- `mot`, le mot à rechercher ;
- `debut` et `fin`, les indices entre lesquels la recherche se fait dans `liste`.

Cette fonction renverra `True` si `liste` contient `mot` entre les indices `debut` (inclus) et `fin` (exclus), et renverra `False` sinon.

Voici son code :

```python
1 def est_present(liste, mot, debut, fin):
2     if debut > fin:
3         return False
4     milieu = (debut + fin) // 2
5     if liste[milieu] > mot:
6         return est_present(liste, mot, 0, milieu - 1)
7     elif liste[milieu] < mot:
8         return est_present(liste, mot, milieu + 1, fin)
9     else:
10        return True
```
Nous testons cette fonction ainsi :
```python
>>> liste_mots = ['FONCTION', 'NSI', 'PYTHON', 'OBJET', 'RAM']
>>> est_present(liste_mots, 'NSI', 0, len(liste_mots))
```

> Dans cette fonction, les entiers `i1` et `i2` représentent respectivement les indices des éléments des listes `L1` et `L2` que l'on souhaite comparer :
> - si aucun des deux indices n'est valide, la boucle `while` est interrompue ;
> - si `i1` n'est plus un indice valide, on va ajouter à `L` les éléments de `L2` à partir de l'indice `i2` ;
> - si `i2` n'est plus un indice valide, on va ajouter à `L` les éléments de `L1` à partir de l'indice `i1` ;
> - sinon, le plus petit élément non encore traité est ajouté à `L` et on décale l'indice correspondant.
>
> Écrire les instructions manquantes des lignes 17 à 22 permettant d'insérer dans la liste `L` les éléments des listes `L1` et `L2` par ordre croissant.

#### ✅ RÉSOLUTION

````python
def fusion(L1, L2):
    L = []
    n1 = len(L1)
    n2 = len(L2)
    i1 = 0
    i2 = 0
    while i1 < n1 or i2 < n2 :
        if i1 >= n1:
            L.append(L2[i2])
            i2 = i2 + 1
        elif i2 >= n2:
            L.append(L1[i1])
            i1 = i1 + 1
        else:
            e1 = L1[i1]
            e2 = L2[i2]
            if e1 <= e2:
                L.append(e1)
                i1 = i1 + 1
            else:
                L.append(e2)
                i2 = i2 + 1
    return L
````

````python
def fusion(liste1, liste2):
    liste_finale = []
    i1, i2 = 0, 0
    while i1 < len(liste1) and i2 < len(liste2):
        if liste1[i1] <= liste2[i2]:
            liste_finale.append(liste1[i1])
            i1 =i1 + 1
        else:
            liste_finale.append(liste2[i2])
            i2 = i2 + 1
    while i1 < len(liste1):
        liste_finale.append(liste1[i1])
        i1 = i1 + 1
    while i2 < len(liste2):
        liste_finale.append(liste2[i2])
        i2 = i2 + 1
    return liste_finale


def tri_fusion(liste):
    if len(liste) <= 1:
        return liste
    milieu = len(liste) // 2
    gauche = tri_fusion(liste[0:milieu])
    droite = tri_fusion(liste[milieu:len(liste)])
    return fusion(gauche, droite)
````

Une fonction `fusionner` a été implémentée en Python en s'inspirant du procédé de la question précédente. Elle prend quatre arguments : la liste qui est en train d'être triée, l'indice où commence la sous-liste de gauche à fusionner, l'indice où termine cette sous-liste, et l'indice où se termine la sous-liste de droite.

*La fonction `fusionner` (le sujet ne donne pas son code ; le voici, à exécuter) :*

````python
def fusionner(liste, i_debut, i_partage, i_fin):
    # fusionne liste[i_debut..i_partage] et liste[i_partage+1..i_fin], déjà triées
    gauche = []
    for k in range(i_debut, i_partage + 1):
        gauche.append(liste[k])
    droite = []
    for k in range(i_partage + 1, i_fin + 1):
        droite.append(liste[k])
    i = 0
    j = 0
    k = i_debut
    while i < len(gauche) and j < len(droite):
        if gauche[i] <= droite[j]:
            liste[k] = gauche[i]
            i = i + 1
        else:
            liste[k] = droite[j]
            j = j + 1
        k = k + 1
    while i < len(gauche):
        liste[k] = gauche[i]
        i = i + 1
        k = k + 1
    while j < len(droite):
        liste[k] = droite[j]
        j = j + 1
        k = k + 1
````

## Exercice 10 — Trouver l'intrus  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique **2024, sujet 40**, exercice 2 (et 2025, sujet 29)

### Question

🏷️ *Type de question :* **D4** Écrire ou compléter la fonction qui divise et s'appelle elle-même · **D5** Couper en deux : moitiés, milieu, indices

#### 📝 ÉNONCÉ

> On considère des tableaux de nombres dont tous les éléments sont présents exactement trois fois à la suite, sauf un élément qui est présent une unique fois et que l'on appelle « l'intrus ». Voici quelques exemples :
> ```python
> tab_a = [3, 3, 3, 9, 9, 9, 1, 1, 1, 7, 2, 2, 2, 4, 4, 4, 8, 8, 8, 5, 5, 5]
> #l'intrus est 7
> tab_b = [8, 5, 5, 5, 9, 9, 9, 18, 18, 18, 3, 3, 3]
> #l'intrus est 8
> tab_c = [5, 5, 5, 1, 1, 1, 0, 0, 0, 6, 6, 6, 3, 8, 8, 8]
> #l'intrus est 3
> ```
> On remarque qu'avec de tels tableaux :
> - pour les indices multiples de 3 situés strictement avant l'intrus, l'élément correspondant et son voisin de droite sont égaux,
> - pour les indices multiples de 3 situés après l'intrus, l'élément correspondant et son voisin de droite - s'il existe - sont différents.
>
> Ce que l'on peut observer ci-dessous en observant les valeurs des paires de voisins marquées par des caractères ^ :
> ```
> [3, 3, 3, 9, 9, 9, 1, 1, 1, 7, 2, 2, 2, 4, 4, 4, 8, 8, 8, 5, 5, 5]
>  ^  ^     ^  ^     ^  ^     ^  ^     ^  ^     ^  ^     ^  ^     ^
>  0        3        6        9        12       15       18       21
> ```
> Dans des listes comme ceux ci-dessus, un algorithme récursif pour trouver l'intrus consiste alors à choisir un indice `i` multiple de 3 situé approximativement au milieu des indices parmi lesquels se trouve l'intrus.
>
> Puis, en fonction des valeurs de l'élément d'indice `i` et de son voisin de droite, à appliquer récursivement l'algorithme à la moitié droite ou à la moitié gauche des indices parmi lesquels se trouve l'intrus.
>
> Par exemple, si on s'intéresse à l'indice 12, on voit les valeurs 2 et 4 qui sont différentes : l'intrus est donc à gauche de l'indice 12 (indice 12 compris)
>
> En revanche, si on s'intéresse à l'indice 3, on voit les valeurs 9 et 9 qui sont identiques : l'intrus est donc à droite des indices 3-4-5, donc à partir de l'indice 6.
>
> Compléter la fonction récursive `trouver_intrus` proposée page suivante qui met en œuvre cet algorithme.
> ```python
> def trouver_intrus(tab, g, d):
>     '''
>     Renvoie la valeur de l'intrus situé entre les indices g et d
>     dans la liste tab où :
>     tab vérifie les conditions de l'exercice,
>     g et d sont des multiples de 3.
>     '''
>     if g == d:
>         return ...
>     else:
>         nombre_de_triplets = (d - g) // ...
>         indice = g + 3 * (nombre_de_triplets // 2)
>         if ... :
>             return ...
>         else:
>             return ...
> ```
> Exemples :
> ```python
> >>> trouver_intrus([3, 3, 3, 9, 9, 9, 1, 1, 1, 7, 2, 2, 2, 4, 4, 4, 8, 8,
> 8, 5, 5, 5], 0, 21)
> 7
> >>> trouver_intrus([8, 5, 5, 5, 9, 9, 9, 18, 18, 18, 3, 3, 3], 0, 12)
> 8
> >>> trouver_intrus([5, 5, 5, 1, 1, 1, 0, 0, 0, 6, 6, 6, 3, 8, 8, 8], 0, 15)
> 3
> ```

#### ✅ RÉSOLUTION

````python
def trouver_intrus(tab, g, d):
    '''
    Renvoie la valeur de l'intrus situé entre les indices g et d
    dans la liste tab où :
    tab vérifie les conditions de l'exercice,
    g et d sont des multiples de 3.
    '''
    if g == d:
        return tab[g]
    else:
        nombre_de_triplets = (d - g) // 3
        indice = g + 3 * (nombre_de_triplets // 2)
        if tab[indice] != tab[indice + 1] :
            return trouver_intrus(tab, g, indice)
        else:
            return trouver_intrus(tab, indice + 3, d)
````

*Les tests de la feuille d'exercices :*

````python
assert trouver_intrus([3, 3, 3, 9, 9, 9, 1, 1, 1, 7, 2, 2, 2, 4, 4, 4, 8, 8, 8, 5, 5, 5], 0, 21) == 7
assert trouver_intrus([8, 5, 5, 5, 9, 9, 9, 18, 18, 18, 3, 3, 3], 0, 12) == 8
assert trouver_intrus([5, 5, 5, 1, 1, 1, 0, 0, 0, 6, 6, 6, 3, 8, 8, 8], 0, 15) == 3
for nb in range(0, 10):
    for position in range(nb + 1):
        t = []
        for k in range(nb + 1):
            if k == position:
                t.append(-1)
            if k < nb:
                for r in range(3):
                    t.append(k)
        assert trouver_intrus(t, 0, len(t) - 1) == -1
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Les deux exemples du texte donnent les deux appels** : valeurs différentes à l'indice 12 → l'intrus est à gauche, indice 12 compris → `trouver_intrus(tab, g, indice)` ; valeurs égales à l'indice 3 → à partir de l'indice 6, c'est-à-dire `indice + 3` → `trouver_intrus(tab, indice + 3, d)`.
- Cas de base : `g == d`, il ne reste qu'une case, c'est l'intrus.
- C'est une **dichotomie** : à chaque appel, on garde à peu près la moitié des triplets.

## Exercice 11 — L'élément majoritaire  ⭐⭐⭐

📚 **Tombé au bac :** épreuve écrite, **Asie 2024, jour 2**, exercice 1

*Cet exercice porte sur la programmation Python (listes, dictionnaires) et la méthode “diviser pour régner”.*

Cet exercice est composé de trois parties indépendantes.

Dans cet exercice, on s'intéresse à des algorithmes pour déterminer, s'il existe, l'élément absolument majoritaire d'une liste.

On dit qu'un élément est absolument majoritaire s'il apparaît dans strictement plus de la moitié des emplacements de la liste.

Par exemple, la liste `[1, 4, 1, 6, 1, 7, 2, 1, 1]` admet 1 comme élément absolument majoritaire, car il apparaît 5 fois sur 9 éléments. Par ailleurs, la liste `[1, 4, 6, 1, 7, 2, 1, 1]` n'admet pas d'élément absolument majoritaire, car celui qui est le plus fréquent est 1, mais il n'apparaît que 4 fois sur 8, ce qui ne fait pas plus que la moitié.

### Question 1 ✍️

🏷️ *Type de question :* **D8** Question sur une autre notion du programme (POO, tuples, dictionnaires...)

#### 📝 ÉNONCÉ

> Déterminer les effectifs possibles d'un élément absolument majoritaire dans une liste de taille 10.

#### ✅ RÉSOLUTION

Il faut apparaître strictement plus de 10/2 = 5 fois : les effectifs possibles sont **6, 7, 8, 9 ou 10**.

#### 💡 EXPLICATION PÉDAGOGIQUE

- « Strictement plus de la moitié » : 5 ne suffit pas pour une liste de 10.

**Partie A : Calcul des effectifs de chaque élément sans dictionnaire**

On peut déterminer l'éventuel élément absolument majoritaire d'une liste en calculant l'effectif de chacun de ses éléments.

### Question 2

🏷️ *Type de question :* **D8** Question sur une autre notion du programme (POO, tuples, dictionnaires...)

#### 📝 ÉNONCÉ

> Écrire une fonction `effectif` qui prend en paramètres une valeur `val` et une liste `lst` et qui renvoie le nombre d'apparitions de `val` dans `lst`. Il ne faut pas utiliser la méthode `count`.

#### ✅ RÉSOLUTION

````python
def effectif(val, lst):
    n = 0
    for x in lst:
        if x == val:
            n = n + 1
    return n
````

*Les tests de la feuille d'exercices :*

#### 💡 EXPLICATION PÉDAGOGIQUE

> ```python
> 1 def eff_dico(lst):

> 2     n = len(lst)

Dans la suite, on suppose qu'on dispose de la fonction `moitie_droite(tab)` qui renvoie la moitié droite sans l'élément du milieu.

*Les fonctions `tri`, `moitie_droite` et `nb_inv_tab` dont on dispose (à exécuter) :*

````python
def tri(tab):
    res = []
    for x in tab:
        res.append(x)
    for i in range(1, len(res)):
        x = res[i]
        j = i - 1
        while j >= 0 and res[j] > x:
            res[j + 1] = res[j]
            j = j - 1
        res[j + 1] = x
    return res


def moitie_droite(tab):
    res = []
    for i in range((len(tab) + 1) // 2, len(tab)):
        res.append(tab[i])
    return res


def nb_inv_tab(tab1, tab2):
    cpt = 0
    for x in tab1:
        for y in tab2:
            if x > y:
                cpt = cpt + 1
    return cpt
````

On donne ci-dessous deux exemples d'appel de cette fonction :
> ```python
> >>> nb_inv_tab([3, 7, 9], [2, 10])
> 3
> >>> nb_inv_tab([7, 9, 13], [7, 10, 14])
> 3
> ```
> En utilisant la fonction `nb_inv_tab` et les questions précédentes, écrire une fonction récursive `nb_inversions_rec(tab)` qui permet de calculer le nombre d'inversions dans un tableau. Cette fonction renverra le même nombre que `nombre_inversions(tab)` de la partie A. On procédera de la façon suivante :
> - Séparer le tableau en deux tableaux de tailles égales (à une unité près).
> - Appeler récursivement la fonction `nb_inversions_rec` pour compter le nombre d'inversions dans chacun des deux tableaux.
> - Trier les deux tableaux (on rappelle qu'une fonction de tri est déjà définie).
> - Ajouter au nombre d'inversions précédemment comptées le nombre renvoyé par la fonction `nb_inv_tab` avec pour arguments les deux tableaux triés.

#### ✅ RÉSOLUTION

````python
def nb_inversions_rec(tab):
    if len(tab) <= 1:
        return 0
    gauche = moitie_gauche(tab)
    droite = moitie_droite(tab)
    total = nb_inversions_rec(gauche) + nb_inversions_rec(droite)
    return total + nb_inv_tab(tri(gauche), tri(droite))
````

Pour trier les éléments situés entre les indices `i` et `j`, où `i < j`, dans un tableau `t` par ce tri, on procède ainsi :
- si les éléments d'indice `i` et `j` sont mal placés, on les échange ;
- si il y a au moins trois éléments entre les indices `i` et `j`
  - on trie les deux premiers tiers du tableau avec cette méthode ;
  - on trie les deux derniers tiers du tableau avec cette méthode ;
  - on trie à nouveau les deux premiers tiers du tableau avec cette méthode.

Pour réaliser ce découpage en tiers, on considère l'entier `k` défini par l'expression `(j – i + 1) // 3`, et on considère les indices intermédiaires `i+k` et `j-k`.

```
                      <------------------------>
                         deux derniers tiers
```

Voici le code partiel de l'algorithme du tri de Stooge en Python qui trie donc les éléments d'un tableau par ordre croissant.
```python
1 def triStooge(tab, i, j):
2     if tab[i] > tab[j]:
3         echange(tab, i, j)
4     if (j - i) > 1:
5         k = (j - i + 1)//3
6         triStooge(...)
7         triStooge(...)
```

### Questions 1 et 2

🏷️ *Type de question :* **D4** Écrire ou compléter la fonction qui divise et s'appelle elle-même

#### 📝 ÉNONCÉ

> 1. Écrire la fonction `echange(tab, i, j)` qui prend en arguments une liste Python `tab` et deux indices `i`, `j`, et réalise sur place l'échange des valeurs dans `tab` à ces indices. La fonction ne renvoie rien.
> 2. Finaliser le programme précédent en complétant les lignes 6, 7 et 8 sur votre copie.

#### ✅ RÉSOLUTION

````python
def echange(tab, i, j):
    temp = tab[i]
    tab[i] = tab[j]
    tab[j] = temp


def triStooge(tab, i, j):
    if tab[i] > tab[j]:
        echange(tab, i, j)
    if (j - i) > 1:
        k = (j - i + 1)//3
        triStooge(tab, i, j - k)
        triStooge(tab, i + k, j)
        triStooge(tab, i, j - k)
````

````python
from math import sqrt


class Segment:
    def __init__(self, point1, point2):
        self.p1 = point1
        self.p2 = point2
        self.longueur = sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)


def liste_segments(liste_points):
    n = len(liste_points)
    segments = []
    for i in range(n - 1):
        for j in range(i + 1, n):
            seg = Segment(liste_points[i], liste_points[j])
            segments.append(seg)
    return segments
````

```python
>>> liste = [1, 2, 3, 4]              >>> liste = [1, 2, 3, 4, 5]

```

*Les fonctions `moitie_gauche` et `moitie_droite` (à exécuter) :*

````python
def moitie_gauche(liste):
    res = []
    for i in range((len(liste) + 1) // 2):
        res.append(liste[i])
    return res


def moitie_droite(liste):
    res = []
    for i in range((len(liste) + 1) // 2, len(liste)):
        res.append(liste[i])
    return res
````

> - Tester si le cas de base est atteint, c'est-à-dire lorsque la liste contient un seul segment ;
> - Découper la liste en deux listes de tailles égales (à une unité près) ;
> - Appeler récursivement la fonction pour rechercher le minimum dans chacune des deux listes ;
> - Comparer les deux valeurs récupérées et renvoyer la plus petite des deux.

#### ✅ RÉSOLUTION

````python
def plus_court_segment(segments):
    if len(segments) == 1:
        return segments[0]
    seg_g = plus_court_segment(moitie_gauche(segments))
    seg_d = plus_court_segment(moitie_droite(segments))
    if seg_g.longueur <= seg_d.longueur:
        return seg_g
    else:
        return seg_d
````

> - Si `haut < long`, on coupe verticalement la grille en deux sous-labyrinthes de même dimension.

#### ✅ RÉSOLUTION

**Le raisonnement (l'arbre des découpes) :**
- 4 × 8 : 4 < 8 → coupe **verticale** en deux grilles 4 × 4, passage tout en haut entre les colonnes 3 et 4.
- chaque 4 × 4 : 4 ≥ 4 → coupe **horizontale** en deux grilles 2 × 4, passage tout à gauche.
- chaque 2 × 4 : coupe verticale en deux 2 × 2, passage en haut.
- chaque 2 × 2 : coupe horizontale en deux 1 × 2 (cas de base : on ouvre le mur du milieu), passage à gauche.

**Le dessin** (obtenu en exécutant l'algorithme ci-dessous) :

````python
def dessiner(lab):
    grille = lab.grille
    haut = len(grille)
    long = len(grille[0])
    for i in range(haut):
        ligne_murs = '+'
        ligne_cases = ''
        for j in range(long):
            if grille[i][j].murs['N']:
                ligne_murs = ligne_murs + '---+'
            else:
                ligne_murs = ligne_murs + '   +'
            if grille[i][j].murs['O']:
                ligne_cases = ligne_cases + '|   '
            else:
                ligne_cases = ligne_cases + '    '
        ligne_cases = ligne_cases + '|'
        print(ligne_murs)
        print(ligne_cases)
    print('+' + '---+' * long)


lab = Labyrinthe(4, 8)
lab.creer_labyrinthe(0, 0, 4, 8)
dessiner(lab)
````

L'algorithme de Douglas-Peuker permet de simplifier une ligne polygonale en supprimant certains de ses sommets (c'est ce qu'on fait pour dessiner le contour d'un pays avec moins de points). ### Question 1 (a, b)

🏷️ *Type de question :* **D8** Question sur une autre notion du programme (POO, tuples, dictionnaires...)

#### 📝 ÉNONCÉ

> La distance qui sépare deux points A et B de coordonnées (x_A, y_A) et (x_B, y_B) est donnée par la formule $AB = \sqrt{(x_B - x_A)^2 + (y_B - y_A)^2}$. On rappelle que la fonction `sqrt` du module `math` de Python renvoie la racine carrée d'un nombre positif ou nul.
>
> **a.** Écrire une instruction qui permet d'importer la fonction `sqrt` du module `math`.
>
> **b.** Supposant l'import réalisé, écrire une fonction `distance_points(a, b)` qui prend en argument deux tuples `a` et `b` représentant les coordonnées de deux points et renvoie la distance qui les sépare.

#### ✅ RÉSOLUTION

````python
from math import sqrt


def distance_points(a, b):
    return sqrt((b[0] - a[0]) ** 2 + (b[1] - a[1]) ** 2)
````

*Les tests de la feuille d'exercices :*

````python
assert distance_points((0, 0), (3, 4)) == 5
assert distance_points((1, 1), (1, 1)) == 0
assert abs(distance_points((-1, 2), (2, -2)) - 5) < 1e-9
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- `a[0]` est l'abscisse de `a`, `a[1]` son ordonnée.
- `** 2` élève au carré. On traduit la formule telle quelle.

*La fonction `distance_point_droite` dont parle la question 2 (à exécuter) :*

````python
def distance_point_droite(p, a, b):
    # distance du point p à la droite (ab) ; erreur si a et b sont égaux
    return abs((b[0] - a[0]) * (a[1] - p[1]) - (a[0] - p[0]) * (b[1] - a[1])) / distance_points(a, b)
````

### Question 2

🏷️ *Type de question :* **D8** Question sur une autre notion du programme (POO, tuples, dictionnaires...)

#### 📝 ÉNONCÉ

> On dispose d'une fonction `distance_point_droite(p, a, b)` qui prend en argument les tuples représentant les coordonnées respectives des points P, A et B, et qui renvoie la distance du point P à la droite (AB). L'exécution de cette fonction produit une erreur dans le cas où les points A et B sont égaux.
>
> À l'aide des fonctions `distance_points` et `distance_point_droite`, écrire une fonction `distance(p, a, b)` qui renvoie la distance entre le point P et la droite (AB) si les points A et B sont distincts et la distance AP sinon.

#### ✅ RÉSOLUTION

````python
def distance(p, a, b):
    if a == b:
        return distance_points(a, p)
    else:
        return distance_point_droite(p, a, b)
````

*Les tests de la feuille d'exercices :*

````python
assert distance((0, 3), (-1, 0), (5, 0)) == 3
assert distance((3, 4), (0, 0), (0, 0)) == 5
assert abs(distance((1, 1), (0, 0), (2, 2))) < 1e-9
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- L'énoncé donne les deux cas : on écrit un `if` / `else`.
- On compare deux tuples avec `==` : ils sont égaux s'ils ont les mêmes coordonnées.

### Question 3

🏷️ *Type de question :* **D8** Question sur une autre notion du programme (POO, tuples, dictionnaires...)

#### 📝 ÉNONCÉ

> Dans la suite, on dira que la fonction `distance` calcule la distance entre le point P et les points A et B, éventuellement confondus.
>
> On a besoin d'une fonction `le_plus_loin(ligne)` qui prend en argument une liste de tuples représentant les coordonnées des points d'une ligne polygonale. Cette fonction doit renvoyer un tuple composé de :
> - l'indice du point de coordonnées `p` de la ligne polygonale d'extrémités `deb` et `fin`, pour lequel la distance `distance(p, deb, fin)` est la plus grande ;
> - la valeur correspondante de cette distance.
>
> On fournit le code incomplet suivant :
> ```python
> def le_plus_loin(ligne):
>     n = len(ligne)
>     deb = ligne[0]
>     fin = ligne[n-1]
>     dmax = 0
>     indice_max = 0
>     for idx in range(1, n-1):
>         p = ...
>         d = distance(p, deb, fin)
>         if ...:
>              ...
>              ...
>     return ...
> ```
> Recopier et compléter le code de cette fonction.

#### ✅ RÉSOLUTION

````python
def le_plus_loin(ligne):
    n = len(ligne)
    deb = ligne[0]
    fin = ligne[n - 1]
    dmax = 0
    indice_max = 0
    for idx in range(1, n - 1):
        p = ligne[idx]
        d = distance(p, deb, fin)
        if d > dmax:
            dmax = d
            indice_max = idx
    return (indice_max, dmax)
````

*Les tests de la feuille d'exercices :*

````python
assert le_plus_loin([(0, 0), (1, 2), (2, 5), (3, 1), (4, 0)]) == (2, 5)
assert le_plus_loin([(0, 0), (1, 0), (2, 0)]) == (0, 0)
assert le_plus_loin([(0, 0), (1, 3), (2, -4), (5, 0)]) == (2, 4)
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est le modèle **« le meilleur »** (recherche du maximum) : on garde le meilleur trouvé jusqu'ici (`dmax`) **et sa position** (`indice_max`), et on les met à jour ensemble quand on trouve mieux.
- `range(1, n - 1)` : on ne regarde que les points **intérieurs** (pas les extrémités).
- On renvoie un **tuple** de deux valeurs : l'appel s'écrira `indice_max, dmax = le_plus_loin(ligne)`.

### Question 4

🏷️ *Type de question :* **D5** Couper en deux : moitiés, milieu, indices

#### 📝 ÉNONCÉ

> Écrire une fonction `extrait(tab, i, j)` qui renvoie la copie du tableau `tab` des cases d'indice `i` inclus à `j` inclus pour 0 ≤ i ≤ j < len(tab). L'appel `extrait([7, 4, 9, 12], 2, 3)` renverra ainsi `[9, 12]`.

#### ✅ RÉSOLUTION

````python
def extrait(tab, i, j):
    res = []
    for k in range(i, j + 1):
        res.append(tab[k])
    return res
````

*Les tests de la feuille d'exercices :*

````python
assert extrait([7, 4, 9, 12], 2, 3) == [9, 12]
assert extrait([7, 4, 9, 12], 0, 0) == [7]
assert extrait([7, 4, 9, 12], 0, 3) == [7, 4, 9, 12]
t = [1, 2, 3]
c = extrait(t, 0, 2)
c.append(4)
assert t == [1, 2, 3]          # c'est bien une copie
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- `j` est **inclus** → `range(i, j + 1)`.
- On construit une **nouvelle** liste : modifier la copie ne modifie pas `tab`.
- (On pourrait écrire `tab[i:j+1]`, mais la boucle montre qu'on sait le faire à la main.)

L'algorithme de Douglas-Peuker repose sur une stratégie de type « diviser pour régner ». Pour éliminer des sommets « proches de l'alignement », un seuil est fixé. Étant donnée une ligne polygonale, le principe de l'algorithme est le suivant :
- si la ligne ne contient qu'un ou deux sommets, l'algorithme se termine ;
- sinon, on considère la droite formée par les extrémités de la ligne (son premier et dernier sommet), et on sélectionne le point le plus éloigné de cette droite (dans le cas où les extrémités sont confondues, on sélectionne le point le plus éloigné de celles-ci) :
  - si la distance entre le point sélectionné et la droite (ou les extrémités lorsqu'elles sont confondues) est inférieure au seuil fixé, on ne conserve que les extrémités de la ligne polygonale ;
  - sinon, on applique l'algorithme de manière récursive aux deux parties de la ligne polygonale formées de la séquence formée du premier sommet jusqu'au sommet sélectionné d'une part et de la séquence formée du sommet sélectionné jusqu'au dernier sommet d'autre part. L'algorithme renvoie alors la concaténation des séquences simplifiées ainsi obtenues.

*Exemple du sujet :* l'algorithme appelé sur la ligne polygonale `[A, B, C, D, E, F, G]` va récursivement être appelé sur les lignes polygonales `[A, B, C]` et `[C, D, E, F, G]`. La ligne polygonale que l'on obtiendra à la fin de l'algorithme sera `[A, C, E, G]`.

#### 📝 ÉNONCÉ

> def simplifie(ligne, seuil):
>     n = len(ligne)

> Recopier et compléter le code de cette fonction.

#### ✅ RÉSOLUTION

````python
def simplifie(ligne, seuil):
    n = len(ligne)
    if n <= 2:
        return ligne
    else:
        indice_max, dmax = le_plus_loin(ligne)
        if dmax <= seuil:
            return [ligne[0], ligne[n - 1]]
        else:
            gauche = simplifie(extrait(ligne, 0, indice_max), seuil)
            droite = simplifie(extrait(ligne, indice_max, n - 1), seuil)
            # le point sélectionné est à la fin de gauche ET au début de droite :
            # on ne le garde qu'une fois
            return gauche + extrait(droite, 1, len(droite) - 1)
````

*Les tests de la feuille d'exercices :*

````python
A, B, C, D = (0, 0), (1, 1.5), (2, 3), (3, 1)
E, F, G = (4, -1), (5, -0.4), (6, 0)
assert simplifie([A, B, C, D, E, F, G], 0.5) == [A, C, E, G]
assert simplifie([A, G], 0.5) == [A, G]
assert simplifie([A, B, C], 0.5) == [A, C]
assert simplifie([A, C, G], 5) == [A, G]
assert simplifie([A, C, G], 1) == [A, C, G]
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base** : 1 ou 2 sommets, rien à simplifier → on renvoie la ligne.
- **Tous les points proches de la droite** → on ne garde que les deux extrémités.
- **Sinon, diviser pour régner** : deux appels récursifs, de `0` à `indice_max` et de `indice_max` à `n - 1` (c'est à ça que sert `extrait`), puis on **combine** en concaténant.
- ⚠️ Le point sélectionné appartient aux **deux** morceaux : si on écrit `gauche + droite`, il apparaît deux fois. Sur l'exemple, on obtiendrait `[A, C, C, E, E, G]` au lieu de `[A, C, E, G]`.
- Pourquoi ça s'arrête : chaque morceau a strictement moins de points que la ligne de départ (le point sélectionné est intérieur).
