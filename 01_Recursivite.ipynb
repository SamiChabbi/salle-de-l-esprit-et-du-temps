### Fiche R5 — Écrire une fonction récursive

🔎 **Je la reconnais :** « Écrire une fonction récursive… », « En utilisant les questions précédentes, écrire une fonction récursive… », ou bien l'énoncé donne une **relation** : `N(m, n) = N(m - 1, n) + N(m, n - 1)`, `d(n) = 2 × d(n - 1) + 1`, `score_max(i, j, p) = p[i][j] + max(...)`.

🛠️ **La méthode des 3 questions** (à écrire au brouillon, en français, **avant** le code) :
1. **Quel est le cas le plus petit, dont je connais la réponse tout de suite ?** → c'est le **cas de base**.
2. **Comment je me ramène à un problème plus petit ?** → `n - 1`, `n // 2`, l'indice suivant `i + 1`, la liste sans son premier élément… → c'est **l'appel récursif**.
3. **Si quelqu'un me donne la réponse du petit problème, comment je fabrique la réponse du grand ?** → c'est le `return`.

Puis on recopie le **squelette** :
```python
def f(parametres):
    if cas_de_base:
        return reponse_directe
    else:
        return ... f(parametres_plus_petits) ...
```
Si l'énoncé donne une formule, **le travail est déjà fait** : la formule est le `return`, et les valeurs données « à part » sont les cas de base.

📌 **Exemple.** Écrire une fonction récursive `puissance(x, n)` qui renvoie x à la puissance n (n entier positif).

| Question | Réponse au brouillon |
|:---|:---|
| 1. Le plus petit cas ? | n = 0 : x⁰ = 1 |
| 2. Plus petit ? | `puissance(x, n - 1)` |
| 3. Comment je fabrique la réponse ? | xⁿ = x × xⁿ⁻¹, donc `x * puissance(x, n - 1)` |

````python
def puissance(x, n):
    if n == 0:
        return 1
    else:
        return x * puissance(x, n - 1)


print(puissance(2, 10))      # 1024
````

⚠️ **Pièges relevés dans la base :**
- oublier le `return` devant l'appel récursif : la fonction calcule… puis renvoie `None` ;
- un cas de base qu'on ne rencontre jamais (avec `n - 2`, un `n` impair saute par-dessus `0` : exercice 4) ;
- oublier un cas de base quand il y a **deux** appels (Fibonacci recule de 1 **et** de 2 : il faut deux cas de base) ;
- avec une liste, utiliser un **indice** qui avance (`i + 1`) est plus simple que de fabriquer des sous-listes ; le cas de base est alors « l'indice est à la fin ».

### Fiche R2 — Prévoir ce qu'affiche ou renvoie un appel (la trace)

🔎 **Je la reconnais :** « Qu'affiche la commande `f(5)` ? », « Écrire ce qui sera affiché… », « Quelle valeur sera affectée à `res` ? », « Combien de fois est appelée la fonction ? (on comptera l'appel initial) », « Donner les affichages produits par… », « Remplir le tableau ».

🛠️ **La méthode : le tableau de trace des appels.** Une ligne par appel, en descendant ; on remonte ensuite pour les valeurs renvoyées.

| Appel | Paramètres | Ce qui est affiché (dans l'ordre) | Ce qui est renvoyé |
|:---|:---|:---|:---|
| 1 | … | … | (rempli en remontant) |

- **On descend** : chaque appel attend la réponse du suivant. On note les `print` **au moment où ils s'exécutent**.
- **On remonte** : à partir du cas de base, on calcule les valeurs renvoyées, de bas en haut.
- **S'il y a deux appels** (Fibonacci, tri fusion, baguenaudier) : on dessine un **arbre des appels**, et on le lit **branche gauche entière d'abord**.

📌 **Exemple.** Que renvoie `f(4)`, et qu'affiche-t-il ?

````python
def f(n):
    if n == 0:
        return 0
    print(n)
    return n + f(n - 2)


res = f(4)
print("res =", res)
````

| Appel | `n` | affiché | renvoyé |
|:---|:---|:---|:---|
| `f(4)` | 4 | 4 | 4 + 2 = **6** ↑ |
| `f(2)` | 2 | 2 | 2 + 0 = 2 ↑ |
| `f(0)` | 0 | rien (cas de base) | 0 ↑ |

Affichage : `4` puis `2`. Valeur : `res = 6`.

⚠️ **Pièges relevés dans la base :**
- un `print` placé **avant** l'appel récursif affiche en descendant (4, 2) ; placé **après**, il affiche en remontant (2, 4) ;
- « combien d'appels » : **on compte l'appel initial** (le sujet le rappelle souvent) ;
- avec un arbre, ne pas lire « étage par étage » : l'appel de gauche est **entièrement fini** avant que celui de droite commence ;
- remplir un tableau de valeurs (`f(1)`, `f(2)`… `f(12)`) : on **réutilise** les cases déjà calculées, on ne redéroule pas tout.

### Fiche R3 — Terminaison : pourquoi ça s'arrête (ou pas)

🔎 **Je la reconnais :** « Expliquer pourquoi le programme s'arrête… », « Justifier que l'appel se terminera quelle que soit… », « L'appel `f2(5)` termine-t-il ? », « Expliquer pourquoi ce cas amènerait une erreur », « Ce code est-il adapté à 2000 cases ? », « Que va-t-il se passer avec n = 10000 ? ».

🛠️ **La méthode :**
1. Repérer la **quantité qui diminue** à chaque appel : `n`, la longueur de la chaîne, `j - i`, le nombre de cases…
2. Vérifier qu'en diminuant, elle **tombe forcément** sur le cas de base (attention aux pas de 2, aux nombres négatifs).
3. Pour montrer que ça **ne** termine **pas**, donner la suite des appels : `f2(5)` → `f2(3)` → `f2(1)` → `f2(-1)` → … on ne passe jamais par 0.

**Phrase type :** « La fonction se termine car à chaque appel, `n` diminue de 1 ; il finit donc par valoir 0, qui est le cas de base, et là la fonction ne s'appelle plus. »

**`RecursionError`** : Python garde en mémoire les appels **en attente** ; au-delà d'environ **1000**, il s'arrête avec `RecursionError: maximum recursion depth exceeded`. Une fonction peut donc être juste en théorie, mais inutilisable pour 2000 ou 10000 (baguenaudier, Fibonacci).

⚠️ **Pièges relevés dans la base :** une boucle `while i != 10` à partir de 12 ne s'arrête jamais (exercice 4, partie A) ; un tirage au hasard (`choice`) qui peut « reculer l'arrêt » indéfiniment (exercice 14).

### Fiche R4 — Compléter une fonction à trous

🔎 **Je la reconnais :** « Recopier et compléter… », « Compléter les lignes 4, 11 et 13 », « Choisir parmi les propositions celle qui convient », du code avec des `...` ou des `à compléter`.

🛠️ **La méthode :**
1. Lire la **docstring et l'exemple** : qu'est-ce que la fonction doit renvoyer ?
2. **Le trou du cas de base** : que renvoie la fonction pour le plus petit cas ? (souvent `0`, `1`, `True`, `[]`, `""`)
3. **Le trou de l'appel** : les paramètres doivent être **plus petits** et du **bon type** (regarder l'appel initial pour voir l'ordre des paramètres).
4. **Vérifier** avec la plus petite valeur qui n'est pas le cas de base.

⚠️ **Piège :** recopier **toute** la fonction sur la copie si le sujet dit « recopier et compléter » ; ne compléter **que** les lignes demandées si le sujet dit « écrire les lignes 6 à 8 ».

### Fiche R6 — Écrire la version itérative (avec une boucle)

🔎 **Je la reconnais :** « Écrire une fonction non récursive… », « Écrire une version itérative… », « Proposer une version itérative… ».

🛠️ **La méthode : la récursive donne déjà tout.**

| Dans la version récursive | Dans la version avec boucle |
|:---|:---|
| la valeur du cas de base | la **valeur de départ** de l'accumulateur |
| l'opération du `return` (`n + ...`, `x * ...`) | ce qu'on fait **à chaque tour** de boucle |
| les appels f(n), f(n − 1), …, f(0) | les tours de boucle (`for k in range(...)`) |

````python
def somme_rec(n):                # récursive
    if n == 0:
        return 0
    return n + somme_rec(n - 1)


def somme_boucle(n):             # itérative : même départ (0), même opération (+)
    total = 0
    for k in range(1, n + 1):
        total = total + k
    return total


print(somme_rec(100), somme_boucle(100))
````

⚠️ **Pièges :** pour un palindrome, on compare `chaine[i]` et `chaine[len(chaine) - 1 - i]` jusqu'au milieu ; pour Fibonacci, il faut **deux** variables qui gardent les deux derniers termes.

### Fiche R7 — Coût, nombre d'appels, calculs en double

🔎 **Je la reconnais :** « Expliquer pourquoi le nombre d'appels augmente aussi vite », « Justifier que tester tous les conduits n'est pas raisonnable », « Quel est le nombre d'appels récursifs ? », « Expliquer comment éviter la redondance des calculs ».

🛠️ **La méthode :** dessiner le début de l'**arbre des appels** et repérer les appels **identiques** (pour `fibonacci(5)`, `fibonacci(3)` est calculé deux fois, `fibonacci(2)` trois fois). Quand chaque appel en fait deux, le nombre d'appels **double à chaque niveau** : de l'ordre de 2ⁿ.

**Les remèdes à citer :** une version avec une boucle, ou garder les résultats déjà calculés dans un tableau ou un dictionnaire (**mémoïsation**, c'est la programmation dynamique).

````python
nb_appels = 0

def fib(n):
    global nb_appels
    nb_appels = nb_appels + 1
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)


for n in [10, 20, 25]:
    nb_appels = 0
    fib(n)
    print("fib(" + str(n) + ") :", nb_appels, "appels")
````

## 5. Les 4 formes de fonctions récursives qu'on rencontre

Toutes les fonctions de la base entrent dans l'une de ces 4 formes. Reconnaître la forme, c'est avoir déjà la moitié du code.

| Forme | Le paramètre qui diminue | Squelette | Dans la base |
|:---|:---|:---|:---|
| **1. Sur un entier** | `n` → `n - 1` ou `n // 2` | `if n == 0: return ...` / `return ... f(n - 1)` | factorielle, somme, compte à rebours, binaire, Syracuse, multiplication |
| **2. Sur une liste ou une chaîne** | un indice `i` → `i + 1` (ou la chaîne raccourcie) | `if i == len(t): return ...` / `return ... t[i] ... f(t, i + 1)` | palindrome, chiffres romains, rendu de monnaie, règles A et B |
| **3. Plusieurs appels** | plusieurs paramètres, plusieurs appels | `return f(...) + f(...)` | Fibonacci, chemins dans une grille, baguenaudier, zone d'une image |
| **4. Un choix à faire** | l'indice de l'objet qu'on envisage | `option1 = f(sans)` / `option2 = f(avec)` / garder la meilleure | charger un camion, forer un puits |

*La forme « couper en deux moitiés » est le chapitre suivant : diviser pour régner.*

### Question 1.a ✍️

🏷️ *Type de question :* **R1** Dire ce qu'est une fonction récursive, justifier qu'une fonction est récursive

#### 📝 ÉNONCÉ

> Expliquer en quelques mots ce qu'est une fonction récursive.

#### ✅ RÉSOLUTION

Une fonction récursive est une fonction **qui s'appelle elle-même**. Elle contient un **cas de base**, où elle renvoie directement un résultat, et un **cas récursif**, où elle s'appelle sur un problème plus petit.

#### 💡 EXPLICATION PÉDAGOGIQUE

- Les deux mots-clés attendus : « **s'appelle elle-même** » et « **cas de base** » (ou condition d'arrêt).

#### 📝 ÉNONCÉ

> En mathématiques, la factorielle d'un entier naturel $n$ est le produit des nombres entiers strictement positifs inférieurs ou égaux à $n$. Par convention, la factorielle de 0 est 1. Par exemple :
> - la factorielle de 1 est 1 ;
> - la factorielle de 2 est 2 × 1 = 2 ;
> - la factorielle de 3 est 3 × 2 × 1 = 6 ;
> - la factorielle de 4 est 4 × 3 × 2 × 1 = 24…
>
> Recopier et compléter le programme ci-dessous afin que la fonction récursive `fact` renvoie la factorielle de l'entier passé en paramètre. Exemple : `fact(4)` renvoie 24.
> ```python
> 1  def fact(n):
> 2      """ Renvoie le produit des nombres entiers
> 3      strictement positifs inférieurs à n """
> 4      if n == 0:
> 5          return à compléter
> 6      else:
> 7          return à compléter
> ```

#### ✅ RÉSOLUTION

````python
def fact(n):
    """ Renvoie le produit des nombres entiers
    strictement positifs inférieurs à n """
    if n == 0:
        return 1
    else:
        return n * fact(n - 1)
````

> def produit(s, n):
>     if n == 0:
>         return [""]

> **a.** Que renvoie la commande `produit("ab", 0)` ? Le résultat est-il une liste vide ?
>
> **b.** Que renvoie la commande `produit("ab", 1)` ?
>
> **c.** Que renvoie la commande `produit("ab", 2)` ?

#### ✅ RÉSOLUTION

**a.** `produit("ab", 0)` renvoie `[""]` (cas de base). Ce n'est **pas** une liste vide : elle contient un élément, la chaîne vide.

**b.** `produit("ab", 1)` : on calcule `produit("ab", 0)`, qui vaut `[""]`, puis
- pour `i = 0` : `ajouter("a", [""])` donne `["a"]` ;
- pour `i = 1` : `ajouter("b", [""])` donne `["b"]`.

**c.** `produit("ab", 2)` : on calcule `produit("ab", 1)`, qui vaut `["a", "b"]`, puis
- pour `i = 0` : `ajouter("a", ["a", "b"])` donne `["aa", "ab"]` ;

````python
def produit(s, n):
    if n == 0:
        return [""]
    else:
        res = []
        for i in range(len(s)):
            res = res + ajouter(s[i], produit(s, n-1))
        return res


print(produit("ab", 0))
print(produit("ab", 1))
print(produit("ab", 2))
````

### Question 1 ✍️

🏷️ *Type de question :* **R4** Compléter une fonction récursive à trous · **R2** Prévoir ce qu'affiche ou renvoie un appel, compter les appels (trace)

#### 📝 ÉNONCÉ

> On donne ci-dessous une implémentation récursive incomplète pour la fonction `palindrome`. L'opération `+` à la ligne 8 permet de concaténer deux chaînes de caractères.
> ```python
> 1  def palindrome(txt):
> 2      """ Str -> Bool """
> 3      ________________ :
> 4          ________________
> 5      taille = len(txt)
> 6      interieur = ""
> 7      for i in range(1, taille - 1):
> 8          interieur = interieur + txt[i]

> ```
> **(a)** Choisir parmi les propositions ci-dessous celle qui convient pour compléter la fonction `palindrome` (lignes 3 et 4) :
>
> | Proposition 1 | Proposition 2 | Proposition 3 | Proposition 4 |
> |:---|:---|:---|:---|
> | `if len(txt) <= 2:`<br>`    return True` | `if len(txt) < 2:`<br>`    return False` | `if len(txt) < 2:`<br>`    return True` | `if len(txt) <= 2:`<br>`    return False` |
>
> **(b)** Lors de l'appel de `palindrome("bonjour")`, indiquer les valeurs, à la ligne 9, de : `txt[0]`, `txt[taille - 1]` et `interieur`.

#### ✅ RÉSOLUTION

**(a)** La **proposition 3** : un mot de 0 ou 1 lettre est un palindrome, il faut donc renvoyer `True` quand `len(txt) < 2`.
- Les propositions 2 et 4 renvoient `False` pour `""` ou `"A"`, qui sont des palindromes.
- La proposition 1 renvoie `True` pour `"AB"`, qui n'est pas un palindrome.

**(b)** `txt[0]` vaut `"b"`, `txt[taille - 1]` vaut `"r"` (`taille` vaut 7), et `interieur` vaut `"onjou"` (les lettres d'indices 1 à 5).

````python
def palindrome(txt):
    """ Str -> Bool """
    if len(txt) < 2:
        return True
    taille = len(txt)
    interieur = ""
    for i in range(1, taille - 1):
        interieur = interieur + txt[i]
    return (txt[0] == txt[taille - 1]) and (palindrome(interieur))


for mot in ["", "A", "AB", "BOB", "LAVAL", "bonjour"]:
    print(repr(mot), palindrome(mot))
````

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `multiplication(n1, n2)` qui renvoie le produit de deux entiers relatifs (positifs ou négatifs) `n1` et `n2`, en n'utilisant **que des additions et des soustractions** (pas de `*`).
>
> Exemples : `multiplication(3, 5)` renvoie `15`, `multiplication(-4, -8)` renvoie `32`, `multiplication(-2, 6)` renvoie `-12`, `multiplication(-2, 0)` renvoie `0`.
>
> *Indice : $n_1 \times n_2 = n_1 + n_1 \times (n_2 - 1)$. Et si `n2` est négatif, $n_1 \times n_2 = -(n_1 \times (-n_2))$.*

#### ✅ RÉSOLUTION

````python
def multiplication(n1, n2):
    if n2 == 0:
        return 0
    if n2 < 0:
        return -multiplication(n1, -n2)       # on se ramène à n2 positif
    return n1 + multiplication(n1, n2 - 1)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** multiplier par `0` donne `0`.
- **Cas récursif (n2 positif) :** $3 \times 5 = 3 + 3 \times 4$ : on ajoute `n1`, et on diminue `n2` de 1.
- **n2 négatif :** si on diminuait `n2`, il s'éloignerait de `0` et la fonction ne s'arrêterait jamais (`RecursionError`). On se ramène donc d'abord au cas positif : $(-2) \times (-3) = -((-2) \times 3)$. Il n'y a pas besoin de traiter `n1` négatif : additionner un nombre négatif marche très bien.
- C'est un bon exemple de la question 3 de la méthode : **vérifier qu'on se rapproche du cas de base**, dans tous les cas.

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `dec_to_bin(n)` qui renvoie l'écriture binaire de l'entier positif `n`, sous forme de chaîne de caractères. Exemple : `dec_to_bin(25)` renvoie `"11001"`.
>
> *Indice : le dernier chiffre binaire de `n` est le reste `n % 2`, et les chiffres d'avant sont l'écriture binaire du quotient `n // 2`. Exemple : $25 = 2 \times 12 + 1$, donc l'écriture de 25 est celle de 12 (`"1100"`) suivie de `"1"`. La fonction `str` transforme un nombre en chaîne : `str(1)` vaut `"1"`.*

#### ✅ RÉSOLUTION

````python
def dec_to_bin(n):
    q = n // 2
    r = n % 2
    if q == 0:                       # n vaut 0 ou 1 : un seul chiffre
        return str(r)
    return dec_to_bin(q) + str(r)
````

*Les tests de la feuille d'exercices :*

````python
assert dec_to_bin(25) == "11001"
assert dec_to_bin(0) == "0"
assert dec_to_bin(1) == "1"
assert dec_to_bin(42) == "101010"
assert dec_to_bin(255) == "11111111"
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est la méthode des **divisions successives par 2** vue en Première, écrite de façon récursive : chaque division donne un chiffre, en partant de la droite.
- Comme le reste `r` est le chiffre **de droite**, on le colle **après** l'écriture du quotient : `dec_to_bin(q) + str(r)`.
- **Cas de base :** quand le quotient vaut `0`, `n` vaut `0` ou `1` : un seul chiffre, `str(r)`.

### Question 2

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `bin_to_dec(mot, i)` qui renvoie la valeur décimale des chiffres binaires `mot[0]`, `mot[1]`, ..., `mot[i]` (de l'indice 0 jusqu'à l'indice `i` inclus). On l'appelle avec `bin_to_dec(mot, len(mot) - 1)` pour tout le mot.
>
> Exemple : `bin_to_dec("101010", 5)` renvoie `42`.
>
> *Indice : ajouter un chiffre binaire à droite **double** la valeur, puis ajoute ce chiffre : `"10101"` vaut 21, donc `"101010"` vaut $2 \times 21 + 0 = 42$.*

#### ✅ RÉSOLUTION

````python
def bin_to_dec(mot, i):
    if mot[i] == "1":                # le chiffre d'indice i, sous forme de nombre
        bit = 1
    else:
        bit = 0
    if i == 0:                       # un seul chiffre
        return bit
    return 2 * bin_to_dec(mot, i - 1) + bit
````

*Les tests de la feuille d'exercices :*

````python
assert bin_to_dec("101010", 5) == 42
assert bin_to_dec("0", 0) == 0
assert bin_to_dec("1", 0) == 1
assert bin_to_dec("11001", 4) == 25
for n in range(300):
    mot = dec_to_bin(n)
    assert bin_to_dec(mot, len(mot) - 1) == n      # les deux fonctions sont inverses
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Ici on avance **de droite à gauche** : `i` part du dernier indice et **diminue**. Cas de base : `i == 0`, il ne reste que le premier chiffre.
- **Cas récursif :** la valeur de `mot[0..i]`, c'est le double de la valeur de `mot[0..i-1]`, plus le chiffre `mot[i]`.
- On vérifie : `"101010"` → $2 \times$ valeur(`"10101"`) $+ 0$ ; valeur(`"10101"`) $= 2 \times$ valeur(`"1010"`) $+ 1$... jusqu'à valeur(`"1"`) $= 1$.

> Pour la suite, il sera utile de disposer d'une fonction `echange` qui permet d'échanger dans une liste `lst` les éléments d'indice `i1` et `i2`. Expliquer pourquoi le code Python ci-dessous ne réalise pas cet échange et en proposer une modification.
> ```python
> def echange(lst, i1, i2):
>     lst[i2] = lst[i1]
>     lst[i1] = lst[i2]
> ```

#### ✅ RÉSOLUTION

````python
def echange(lst, i1, i2):
    temp = lst[i1]
    lst[i1] = lst[i2]
    lst[i2] = temp
````

#### 📝 ÉNONCÉ

> Proposer une version itérative du mélange de Fisher-Yates.

#### ✅ RÉSOLUTION

````python
from random import randint


def melange_iteratif(lst):
    for ind in range(len(lst) - 1, 0, -1):
        j = randint(0, ind)
        echange(lst, ind, j)
````

## Exercice 11 — La suite de Fibonacci  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 28 (exercice 28.1) : `fibonacci(n)`, avec par exemple `fibonacci(25) = 75025`.

La suite de Fibonacci commence par `1, 1`, puis chaque terme est la **somme des deux précédents** : `1, 1, 2, 3, 5, 8, 13, 21, ...`

On note `fibonacci(n)` le terme d'indice `n`, avec `fibonacci(1) = 1` et `fibonacci(2) = 1`.

### Question 1

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `fibonacci(n)` (pour `n >= 1`).

#### ✅ RÉSOLUTION

````python
def fibonacci(n):
    if n == 1 or n == 2:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)
````

*Les tests de la feuille d'exercices :*

````python
assert fibonacci(1) == 1
assert fibonacci(2) == 1
assert fibonacci(6) == 8
assert fibonacci(10) == 55
assert fibonacci(25) == 75025
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La définition est elle-même récursive : il suffit de la recopier.
- Il faut **deux** cas de base, car la fonction a besoin des **deux** termes précédents : avec seulement `n == 1`, l'appel `fibonacci(2)` appellerait `fibonacci(0)`, puis `fibonacci(-1)`... sans jamais s'arrêter.

### Question 2 ✍️

🏷️ *Type de question :* **R7** Coût, nombre d'appels, calculs en double

#### 📝 ÉNONCÉ

> On compte le nombre d'appels avec une variable globale `nb_appels`. Exécuter la cellule ci-dessous, puis expliquer pourquoi le nombre d'appels augmente aussi vite. Essayer `fibonacci_compte(35)` : que se passe-t-il ?

#### ✅ RÉSOLUTION

**Explication :** chaque appel en déclenche **deux** autres, et les mêmes calculs sont refaits un très grand nombre de fois. Par exemple `fibonacci(5)` calcule `fibonacci(4)` et `fibonacci(3)`, mais `fibonacci(4)` recalcule **lui aussi** `fibonacci(3)` :
```
                    fib(5)
                 /          \
            fib(4)          fib(3)
           /      \         /     \
       fib(3)   fib(2)   fib(2)  fib(1)
       /    \
   fib(2)  fib(1)
```
Le nombre d'appels est multiplié par environ 1,6 à chaque fois que `n` augmente de 1 : pour `n = 35`, il faut presque 30 millions d'appels, et le calcul prend plusieurs secondes. Pour `n = 50`, il faudrait des heures.

````python
nb_appels = 0

def fibonacci_compte(n):
    global nb_appels
    nb_appels = nb_appels + 1
    if n == 1 or n == 2:
        return 1
    return fibonacci_compte(n - 1) + fibonacci_compte(n - 2)


for n in [5, 10, 15, 20, 25, 30]:
    nb_appels = 0
    resultat = fibonacci_compte(n)
    print("fibonacci(" + str(n) + ") =", resultat, "  appels :", nb_appels)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est **le** piège classique de la récursivité : deux appels récursifs **sur des sous-problèmes qui se chevauchent** refont sans cesse le même travail.
- Comparaison avec l'exponentiation rapide : là aussi, la « version maladroite » `power_dac(x, n // 2) * power_dac(x, n // 2)` calculait deux fois la même chose ; il suffisait de stocker le résultat dans une variable. Ici, les deux appels sont **différents** (`n - 1` et `n - 2`), donc ce n'est pas aussi simple.
- **Solutions :** une simple **boucle** qui garde les deux derniers termes (question 3), ou la **mémoïsation** (retenir dans un dictionnaire les résultats déjà calculés), qu'on voit en Terminale avec la programmation dynamique.

### Question 3

🏷️ *Type de question :* **R6** Écrire la version itérative (avec une boucle)

#### 📝 ÉNONCÉ

> Écrire une fonction **non récursive** `fibonacci_boucle(n)`, avec une boucle, qui calcule le même résultat très rapidement, même pour `n = 100`.
>
> *Indice : garder dans deux variables les deux derniers termes calculés.*

#### ✅ RÉSOLUTION

````python
def fibonacci_boucle(n):
    avant_dernier = 1
    dernier = 1
    for k in range(n - 2):
        suivant = avant_dernier + dernier
        avant_dernier = dernier
        dernier = suivant
    return dernier


print(fibonacci_boucle(100))
````

*Les tests de la feuille d'exercices :*

````python
for n in range(1, 21):
    assert fibonacci_boucle(n) == fibonacci(n)
assert fibonacci_boucle(100) == 354224848179261915075
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La boucle calcule chaque terme **une seule fois**, en partant du début : `n - 2` additions seulement.
- La récursivité n'est pas toujours la meilleure solution : elle est élégante, mais il faut **réfléchir au nombre d'appels**.

## Exercice 12 — Rendu de monnaie  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 36 (exercice 36.2) : `rendu_glouton(a_rendre, rang)`, avec par exemple `rendu_glouton(67, 0)` qui renvoie `[50, 10, 5, 2]`.

On dispose de pièces et de billets dont les valeurs sont rangées dans la liste `valeurs` (par ordre décroissant). Pour rendre une somme, on utilise l'algorithme **glouton** : on rend toujours la **plus grande** valeur possible, tant qu'elle ne dépasse pas ce qu'il reste à rendre.

```python
valeurs = [100, 50, 20, 10, 5, 2, 1]
```

*Rappel : `+` colle aussi deux listes : `[50] + [10, 5]` donne `[50, 10, 5]`.*

### Question 1

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `rendu_glouton(a_rendre, rang)` qui renvoie la liste des pièces et billets à rendre pour la somme `a_rendre`, en n'utilisant que les valeurs `valeurs[rang]`, `valeurs[rang + 1]`, ... On l'appelle avec `rang = 0`.
>
> *Indice : si `valeurs[rang]` ne dépasse pas `a_rendre`, on la rend, et il reste `a_rendre - valeurs[rang]` à rendre (avec la même valeur possible encore une fois). Sinon, on passe à la valeur suivante (`rang + 1`).*

#### ✅ RÉSOLUTION

````python
valeurs = [100, 50, 20, 10, 5, 2, 1]


def rendu_glouton(a_rendre, rang):
    if a_rendre == 0:
        return []
    v = valeurs[rang]
    if v <= a_rendre:
        return [v] + rendu_glouton(a_rendre - v, rang)      # on rend v, on garde le même rang
    return rendu_glouton(a_rendre, rang + 1)                # v est trop grande : on passe à la suivante
````

*Les tests de la feuille d'exercices :*

````python
assert rendu_glouton(67, 0) == [50, 10, 5, 2]
assert rendu_glouton(291, 0) == [100, 100, 50, 20, 20, 1]
assert rendu_glouton(0, 0) == []
assert rendu_glouton(8, 0) == [5, 2, 1]
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** il n'y a plus rien à rendre : la liste vide `[]`.
- **Deux cas récursifs :**
  - la valeur `v` convient : on la rend (`[v] + ...`) et on rend le reste, **sans changer de rang** (on peut avoir besoin de plusieurs fois la même valeur, comme deux billets de 100) ;
  - elle est trop grande : on ne rend rien et on passe à la valeur suivante.
- **Terminaison :** à chaque appel, soit `a_rendre` diminue, soit `rang` augmente. Comme la dernière valeur est `1`, on finit toujours par tout rendre avant de dépasser la fin de la liste.

## Exercice 13 — Les chiffres romains  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 30 (exercice 30.2) : `traduire_romain(nombre)`. Le sujet utilisait le slicing ; ici on utilise un indice.

En chiffres romains, `I = 1`, `V = 5`, `X = 10`, `L = 50`, `C = 100`, `D = 500`, `M = 1000`. On les stocke dans un dictionnaire :
```python
romains = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}
```
On lit un nombre de gauche à droite. Un symbole **ajoute** sa valeur, **sauf** s'il est suivi d'un symbole de valeur plus grande : dans ce cas, il la **retranche**. Exemples : `"XVI"` vaut $10 + 5 + 1 = 16$, mais `"XIV"` vaut $10 + (5 - 1) = 14$, et `"CXLII"` vaut $100 + (50 - 10) + 1 + 1 = 142$.

### Question 1

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `traduire_romain(nombre, i)` qui renvoie la valeur de la partie de la chaîne `nombre` qui commence à l'indice `i`. On l'appelle avec `traduire_romain(nombre, 0)`.
>
> *Indice : cas de base quand `i` est l'indice du dernier symbole. Sinon, comparer la valeur du symbole `nombre[i]` à celle du suivant `nombre[i + 1]`.*

#### ✅ RÉSOLUTION

````python
romains = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}


def traduire_romain(nombre, i):
    valeur = romains[nombre[i]]
    if i == len(nombre) - 1:                         # dernier symbole
        return valeur
    if valeur >= romains[nombre[i + 1]]:
        return valeur + traduire_romain(nombre, i + 1)
    return traduire_romain(nombre, i + 1) - valeur
````

*Les tests de la feuille d'exercices :*

````python
assert traduire_romain("XIV", 0) == 14
assert traduire_romain("CXLII", 0) == 142
assert traduire_romain("MMXXIV", 0) == 2024
assert traduire_romain("MCMXCIX", 0) == 1999
assert traduire_romain("I", 0) == 1
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** le dernier symbole compte toujours positivement (il n'est suivi de rien).
- **Cas récursif :** on traduit la suite de la chaîne (appel avec `i + 1`), puis on **ajoute** ou on **retranche** la valeur du symbole courant selon le symbole qui le suit.
- Exemple : `"XIV"` → `10 + traduire("IV")` ; `"IV"` → `traduire("V") - 1 = 4` ; total `14`.

#### ✅ RÉSOLUTION

> def regleA(chaine):
>     n = len(chaine)
>     if n >= 2:
>         return chaine[0] == "a" and chaine[n - 1] == "a" and regleA(...)

> (b) Écrire une fonction `regleB(chaine)` qui renvoie `True` si la chaîne suit la règle B, et `False` sinon.

#### ✅ RÉSOLUTION

````python
def raccourcir(chaine):
    res = ""
    for k in range(1, len(chaine) - 1):
        res = res + chaine[k]
    return res


def regleA(chaine):
    n = len(chaine)
    if n >= 2:
        return chaine[0] == "a" and chaine[n - 1] == "a" and regleA(raccourcir(chaine))
    else:
        return chaine == "a"


def regleB(chaine):
    n = len(chaine)
    if n >= 2:
        milieu = raccourcir(chaine)
        return chaine[0] == "b" and chaine[n - 1] == "b" and (regleA(milieu) or regleB(milieu))
    else:
        return False
````

- **Cas de base de `regleA` :** une chaîne de moins de 2 caractères suit la règle A seulement si c'est `"a"`.
- **Cas de base de `regleB` :** une chaîne de moins de 2 caractères ne peut jamais suivre la règle B (elle commence et finit par `"b"` autour d'une autre chaîne non vide).
- `regleB` appelle `regleA` **et** `regleB` : la règle B fait référence aux deux règles.

## Exercice 15 — Chemins dans une grille  ⭐⭐⭐

📚 **Tombé au bac :** épreuve écrite 2024, Centres étrangers 1, jour 2, exercice 1 (partie C, questions 7 à 9).

On se déplace dans une grille rectangulaire de `m` colonnes et `n` lignes, de la case **en haut à gauche** (S) à la case **en bas à droite** (E). Les seuls déplacements autorisés sont **une case vers la droite** (D) ou **une case vers le bas** (B).

```
S → → ↓          un chemin possible dans une grille 4 × 3 : D D B D B
    · ↓ → ↓
    · · · E
```

On note `N(m, n)` le nombre de chemins différents dans une grille de `m` colonnes et `n` lignes.

### Question 1 ✍️

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> Pour une grille d'une seule ligne (dimension `m × 1`), justifier qu'il y a un seul chemin : `N(m, 1) = 1`. De même, `N(1, n) = 1`.

#### ✅ RÉSOLUTION

Dans une grille d'une seule ligne, on ne peut jamais descendre : le seul chemin consiste à aller tout droit vers la droite jusqu'à E. Il y a donc un seul chemin. De même, dans une grille d'une seule colonne, on ne peut que descendre.

#### 💡 EXPLICATION PÉDAGOGIQUE

- Ce sont les **cas de base** de la future fonction récursive.

### Question 2 ✍️

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> Justifier que `N(m, n) = N(m - 1, n) + N(m, n - 1)`.

#### ✅ RÉSOLUTION

Depuis la case de départ, le premier déplacement est soit **vers la droite**, soit **vers le bas** :
- si on va à droite, il reste à aller de cette nouvelle case jusqu'à E : c'est un chemin dans une grille d'**une colonne de moins**, il y en a `N(m - 1, n)` ;
- si on descend, il reste un chemin dans une grille d'**une ligne de moins** : il y en a `N(m, n - 1)`.

Ces deux familles de chemins sont différentes (le premier pas n'est pas le même) et elles contiennent tous les chemins. Donc `N(m, n) = N(m - 1, n) + N(m, n - 1)`.

#### 💡 EXPLICATION PÉDAGOGIQUE

- Raisonnement typique : on **sépare en cas** selon le **premier** choix, et chaque cas est **le même problème en plus petit**.
- On peut aussi raisonner sur le **dernier** déplacement (on arrive en E par la gauche ou par le haut) : on obtient la même formule.

### Question 3

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> En utilisant les questions précédentes, écrire une fonction récursive `nombre_chemins(m, n)` qui renvoie le nombre de chemins possibles dans une grille de dimension `m × n`.

#### ✅ RÉSOLUTION

````python
def nombre_chemins(m, n):
    if m == 1 or n == 1:
        return 1
    return nombre_chemins(m - 1, n) + nombre_chemins(m, n - 1)
````

````python
assert nombre_chemins(1, 5) == 1
assert nombre_chemins(5, 1) == 1
assert nombre_chemins(2, 2) == 2
assert nombre_chemins(3, 3) == 6
assert nombre_chemins(4, 3) == 10
assert nombre_chemins(3, 4) == 10
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Une traduction directe : cas de base (question 1), puis formule (question 2).
- **Terminaison :** à chaque appel, `m` ou `n` diminue de 1, et on s'arrête dès que l'un des deux vaut 1.
- Comme pour Fibonacci, il y a **deux** appels récursifs, et les mêmes grilles sont recalculées plusieurs fois : `nombre_chemins(15, 15)` fait déjà des dizaines de millions d'appels. (Pour aller plus loin : le résultat est le coefficient binomial $\binom{m+n-2}{m-1}$, qu'on retrouve dans le triangle de Pascal.)

#### ✅ RÉSOLUTION

````python
def vider(n):
    if n == 1:
        print('Vider case 1')
    elif n > 1:
        vider(n-2)
        print('Vider case', n)
        remplir(n-2)
        vider(n-1)


def remplir(n):
    if n == 1:
        print('Remplir case 1')
    elif n > 1:
        remplir(n-1)
        vider(n-2)
        print('Remplir case', n)
        remplir(n-2)
````

*Les tests de la feuille d'exercices :*

````python
# On simule la réglette pour vérifier que chaque coup affiché est autorisé
import builtins
reglette = []
affichage = builtins.print


def joue(texte, numero=None):
    if numero is None:
        case = int(texte.split()[-1])
    else:
        case = numero
    premiere = None
    for k in range(len(reglette)):
        if reglette[k] and premiere is None:
            premiere = k
    assert case == 1 or (premiere is not None and case - 1 == premiere + 1), "coup interdit"
    reglette[case - 1] = not reglette[case - 1]


for taille in range(1, 8):
    reglette = []
    for k in range(taille):
        reglette.append(False)
    builtins.print = joue
    try:
        remplir(taille)
    finally:
        builtins.print = affichage
    for case in reglette:
        assert case == True, "toutes les cases doivent être remplies"
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

  3. on peut alors **remplir la case `n`** (elle suit la première case remplie) ;
  4. `remplir(n-2)` : on remplit de nouveau les cases 1 à `n - 2`.
- `vider` et `remplir` s'appellent **l'une l'autre** : c'est une récursivité **croisée**.

Un appel à cette fonction doit permettre de calculer la charge optimale pour un transport de volume `v_restant` utilisant les marchandises à partir de l'indice `i` :
> - si `i >= n`, toutes les marchandises ont été essayées et il n'en reste plus d'autres disponibles. L'appel récursif renvoie la liste vide ;
> - si `i < n` et la marchandise d'indice `i` est de volume strictement supérieur au volume restant, l'appel récursif renvoie le résultat de l'appel effectué avec le même volume restant mais avec la marchandise suivante, c'est-à-dire `chargeOptimale(tab, v_restant, i+1)` ;
> - si `i < n` et la marchandise d'indice `i` est de volume inférieur ou égal au volume restant, il existe deux options possibles :
>   - soit on utilise la marchandise `i`, auquel cas le chargement contiendra cette marchandise et celles du résultat de l'appel récursif à partir de la prochaine marchandise et d'un volume restant strictement inférieur ;
>   - soit on n'utilise pas la marchandise `i`, auquel cas le chargement sera le résultat de l'appel récursif avec le même volume restant mais à partir de la marchandise suivante.
>
>   On garde l'option de chargement qui maximise le prix transporté.
>
> Compléter le code de la fonction `chargeOptimale` dont le principe a été décrit ci-avant.
> ```python
> def chargeOptimale(tab, v_restant, i):
>     if i>=...:
>         return ...
>     else:
>         if tab[i].volume > v_restant:
>             return chargeOptimale(tab, v_restant, i+1)
>         else:
>             option1 = chargeOptimale(tab, ..., ...)
>             option2 = [tab[i]] + chargeOptimale(tab, ..., ...)
>             if prixListe(option1) > prixListe(option2):
>                 return ...
>             else:
>                 return ...
> ```

#### ✅ RÉSOLUTION

````python
class Marchandise:
    def __init__(self, p, v):
        assert v > 0
        self.prix = p
        self.volume = v


def prixListe(tab):
    total = 0
    for m in tab:
        total = total + m.prix
    return total


def chargeOptimale(tab, v_restant, i):
    if i>=len(tab):
        return []
    else:
        if tab[i].volume > v_restant:
            return chargeOptimale(tab, v_restant, i+1)
        else:
            option1 = chargeOptimale(tab, v_restant, i+1)
            option2 = [tab[i]] + chargeOptimale(tab, v_restant - tab[i].volume, i+1)
            if prixListe(option1) > prixListe(option2):
                return option1
            else:
                return option2
````

*Les tests de la feuille d'exercices :*

````python
m1 = Marchandise(40, 20)
m2 = Marchandise(210, 70)
m3 = Marchandise(160, 40)
m4 = Marchandise(50, 50)
tab = [m3, m2, m1, m4]
charge = chargeOptimale(tab, 100, 0)
assert prixListe(charge) == 250
assert m1 in charge and m2 in charge
assert chargeOptimale(tab, 10, 0) == []
assert prixListe(chargeOptimale(tab, 1000, 0)) == 460
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- ⚠️ Dans le code du sujet, `option1` est l'option **sans** la marchandise `i`, et `option2` celle **avec** (on le voit au `[tab[i]] + …`). C'est l'inverse de l'ordre du texte : il faut lire le code.
- Si on prend la marchandise `i`, le volume restant **diminue** de son volume : `v_restant - tab[i].volume`.
- Dans les deux options, on passe à la marchandise suivante : `i+1`. C'est ce qui garantit que la fonction se termine (`i` augmente jusqu'à `len(tab)`).
- Chaque appel en fait deux : le nombre d'appels **double** à chaque marchandise ($2^n$ en tout), d'où un coût exponentiel. La programmation dynamique permet de faire mieux (exercice suivant).

## Exercice 18 — Remplir une zone dans une image  ⭐⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 27 : `colore_comp1(M, i, j, val)`, qui donne la valeur `val` à tous les pixels de la zone du pixel `(i, j)`.

Une image en noir et blanc est une matrice `M` dont les cases (les **pixels**) valent `0` (blanc) ou `1` (noir). Deux pixels noirs sont **voisins** s'ils se touchent par un côté (en haut, en bas, à gauche ou à droite). Une **zone** (on dit aussi une *composante connexe*) est un ensemble de pixels noirs reliés entre eux de voisin en voisin.

C'est l'outil « pot de peinture » des logiciels de dessin : on clique sur un pixel, et toute sa zone change de couleur.

La fonction `afficher` est donnée.

### Question 1

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `colorer(M, i, j, val)` qui donne la valeur `val` (un entier supérieur ou égal à 2) à tous les pixels de la zone qui contient le pixel `(i, j)`. Si `(i, j)` est en dehors de l'image, ou si ce n'est pas un pixel noir (`1`), la fonction ne fait rien. La fonction **modifie** `M` et ne renvoie rien.
>
> *Indice : on colore le pixel `(i, j)`, puis on appelle `colorer` sur ses 4 voisins.*

#### ✅ RÉSOLUTION

````python
def afficher(M):
    for ligne in M:
        for x in ligne:
            print(x, end=" ")
        print()
    print()


def colorer(M, i, j, val):
    if i < 0 or i >= len(M) or j < 0 or j >= len(M[0]):
        return                       # en dehors de l'image
    if M[i][j] != 1:
        return                       # pixel blanc, ou déjà coloré
    M[i][j] = val
    colorer(M, i - 1, j, val)        # en haut
    colorer(M, i + 1, j, val)        # en bas
    colorer(M, i, j - 1, val)        # à gauche
    colorer(M, i, j + 1, val)        # à droite
````

````python
M = [[1, 1, 0, 0, 1],
     [0, 1, 0, 1, 1],
     [0, 1, 0, 0, 0],
     [1, 0, 0, 1, 1]]
afficher(M)
colorer(M, 0, 0, 2)
afficher(M)
````

*Les tests de la feuille d'exercices :*

````python
M = [[1, 1, 0, 0, 1],
     [0, 1, 0, 1, 1],
     [0, 1, 0, 0, 0],
     [1, 0, 0, 1, 1]]
colorer(M, 0, 0, 2)
assert M == [[2, 2, 0, 0, 1],
             [0, 2, 0, 1, 1],
             [0, 2, 0, 0, 0],
             [1, 0, 0, 1, 1]]
colorer(M, 0, 2, 3)              # pixel blanc : rien ne change
colorer(M, 9, 9, 3)              # en dehors : rien ne change
assert M[0][2] == 0
colorer(M, 1, 4, 3)
assert M[0][4] == 3 and M[1][3] == 3 and M[1][4] == 3 and M[3][3] == 1
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Deux cas de base :** on est sorti de l'image, ou le pixel n'est pas noir. Dans les deux cas, on s'arrête sans rien faire (`return` seul).
- **Le point crucial :** on colore le pixel **avant** de lancer les appels sur les voisins. Comme un pixel coloré ne vaut plus `1`, on ne le traitera **pas deux fois**. Sinon, deux voisins s'appelleraient l'un l'autre à l'infini.
- C'est pour ça que `val` doit être différent de `1` (et de `0`) : on prend `2`, `3`...
- **Avec des boucles**, ce problème serait très difficile à écrire. C'est un exemple où la récursivité est vraiment l'outil naturel.

### Question 2

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> En utilisant `colorer`, écrire une fonction `nombre_zones(M)` qui renvoie le nombre de zones noires de l'image. On pourra colorer chaque nouvelle zone rencontrée avec une valeur différente : `2`, puis `3`, puis `4`...

#### ✅ RÉSOLUTION

````python
def nombre_zones(M):
    val = 2
    for i in range(len(M)):
        for j in range(len(M[0])):
            if M[i][j] == 1:                 # pixel noir pas encore coloré : nouvelle zone
                colorer(M, i, j, val)
                val = val + 1
    return val - 2
````

*Les tests de la feuille d'exercices :*

````python
M = [[1, 1, 0, 0, 1],
     [0, 1, 0, 1, 1],
     [0, 1, 0, 0, 0],
     [1, 0, 0, 1, 1]]
assert nombre_zones(M) == 4
afficher(M)
assert nombre_zones([[0, 0], [0, 0]]) == 0
assert nombre_zones([[1, 1], [1, 1]]) == 1
assert nombre_zones([[1, 0, 1], [0, 1, 0], [1, 0, 1]]) == 5    # les diagonales ne comptent pas
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- On parcourt toute l'image avec deux boucles. Chaque fois qu'on trouve un pixel encore noir (`1`), c'est qu'il appartient à une **nouvelle** zone (sinon il aurait déjà été coloré) : on colore toute cette zone d'un coup, et on compte une zone de plus.
- On mélange ici **boucles** (pour parcourir l'image) et **récursivité** (pour remplir une zone) : chaque outil là où il est le plus simple.
- À la fin, `M` a été modifiée : chaque zone a sa propre couleur, comme dans le sujet du bac.

### Question 6

🏷️ *Type de question :* **R5** Écrire une fonction récursive à partir d'une règle ou d'une formule

#### 📝 ÉNONCÉ

> On dira qu'un conduit est **maximal** si son score de confiance est maximal. On observe que :
> - si on a des conduits maximaux pour les deux petites pyramides obtenues en enlevant le sommet, on obtient un conduit maximal en ajoutant le sommet devant le conduit de plus grand score parmi les deux. Par exemple, pour `ex1`, la petite pyramide de gauche (de sommet 6) a un conduit maximal de score 6 + 5 + 6 + 5 = 22, celle de droite (de sommet 2) un conduit maximal de score 2 + 7 + 6 + 5 = 20 : le score maximal pour `ex1` est donc 4 + 22 = 26 ;
> - si la pyramide n'a qu'un seul niveau, le seul conduit possible contient le sommet, et le nombre de ce sommet est le score maximal.
>
> On note `score_max(i, j, p)` le score maximal possible depuis le nombre d'indice `j` du niveau `i` de la pyramide `p`. On a alors les relations suivantes :
> - `score_max(len(p)-1, j, p) = p[len(p)-1][j]` ;
> - `score_max(i, j, p) = p[i][j] + max(score_max(i+1, j, p), score_max(i+1, j+1, p))`.
>
> Le score maximal possible pour `p` toute entière sera alors `score_max(0, 0, p)`.
>
> Écrire la fonction récursive `score_max` qui implémente les règles précédentes.

#### ✅ RÉSOLUTION

````python
def score_max(i, j, p):
    if i == len(p) - 1:
        return p[i][j]
    return p[i][j] + max(score_max(i + 1, j, p), score_max(i + 1, j + 1, p))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Les deux relations de l'énoncé se traduisent **mot pour mot** : la première est le cas de base (dernier niveau), la deuxième le cas récursif.
- Deux appels récursifs par appel : comme pour Fibonacci, on recalcule plusieurs fois les mêmes cases (par exemple `score_max(3, 1, p)` est calculé depuis `score_max(2, 0, p)` **et** depuis `score_max(2, 1, p)`).

### Questions 7 et 8 — programmation dynamique

🏷️ *Type de question :* **R6** Écrire la version itérative (avec une boucle)

#### 📝 ÉNONCÉ

> Pour éviter la redondance des calculs, on construit une pyramide `s` dont le nombre à l'indice `j` du niveau `i` correspond à `score_max(i, j, p)`.
>
> 7. Écrire une fonction `pyramide_nulle` qui prend en paramètre un entier `n` et construit une pyramide remplie de 0 à `n` niveaux.
> 8. Compléter la fonction `prog_dyn` ci-dessous qui prend en paramètre une pyramide `p`, et qui renvoie le score maximal pour un conduit dans `p`. Pour cela, on construit une pyramide `s` remplie de 0 de la même taille et on la remplit avec les valeurs de `score_max` en commençant par le dernier niveau et en appliquant petit à petit les relations données ci-dessus.
> ```python
> 1  def prog_dyn(p):
> 2      n = len(p)
> 3      s = ...
> 4      # remplissage du dernier niveau
> 5      for j in ...
> 6          s[n-1][j] = ...
> 7      # remplissage des autres niveaux
> 8      for i in ...
> 9          for j in ...
> 10             s[i][j] = ...
> 11     # renvoie du score maximal
> 12     return s[0][0]
> ```

#### ✅ RÉSOLUTION

````python
def pyramide_nulle(n):
    p = []
    for i in range(n):
        niveau = []
        for j in range(i + 1):
            niveau.append(0)
        p.append(niveau)
    return p


def prog_dyn(p):
    n = len(p)
    s = pyramide_nulle(n)
    # remplissage du dernier niveau
    for j in range(n):
        s[n-1][j] = p[n-1][j]
    # remplissage des autres niveaux
    for i in range(n - 2, -1, -1):
        for j in range(i + 1):
            s[i][j] = p[i][j] + max(s[i+1][j], s[i+1][j+1])
    # renvoie du score maximal
    return s[0][0]
````

*Les tests de la feuille d'exercices :*

````python
assert pyramide_nulle(1) == [[0]]
assert pyramide_nulle(3) == [[0], [0, 0], [0, 0, 0]]
ex1 = [[4], [6, 2], [3, 5, 7], [5, 1, 6, 2], [4, 7, 3, 5, 2]]
ex2 = [[3], [1, 2], [4, 5, 9], [3, 6, 2, 1]]
assert prog_dyn(ex1) == 26
assert prog_dyn(ex2) == 16
assert prog_dyn([[7]]) == 7
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **`pyramide_nulle` :** le niveau `i` contient `i + 1` cases. On construit chaque niveau avec une boucle, puis on l'ajoute à la pyramide.
- **`prog_dyn` :** on remplit `s` **du bas vers le haut** (`range(n - 2, -1, -1)` donne `n - 2`, …, `0`). Quand on calcule `s[i][j]`, les deux cases du dessous sont déjà connues : plus besoin d'appel récursif, chaque case est calculée **une seule fois**.
- C'est la **remontée** du tableau de trace, faite directement avec des boucles.

### Questions 9 et 10 ✍️

🏷️ *Type de question :* **R7** Coût, nombre d'appels, calculs en double

#### 📝 ÉNONCÉ

> 9. Montrer que le coût d'exécution de la fonction `prog_dyn` est quadratique en `n` pour une pyramide à `n` niveaux.
> 10. Expliquer comment adapter la fonction `score_max` pour éviter la redondance des calculs afin d'obtenir également un coût quadratique, tout en gardant une approche récursive.

#### ✅ RÉSOLUTION

9. `prog_dyn` calcule chaque case de la pyramide **une seule fois**, en temps constant. La pyramide contient $1 + 2 + \cdots + n = \dfrac{n(n+1)}{2}$ cases : le coût est de l'ordre de $n^2$, donc **quadratique**.

10. On utilise la **mémoïsation** : on garde en mémoire (par exemple dans un dictionnaire, dont les clés sont les couples `(i, j)`) les valeurs de `score_max` déjà calculées. Au début de chaque appel, si la valeur de `(i, j)` est déjà dans le dictionnaire, on la renvoie directement sans refaire d'appel récursif ; sinon, on la calcule, on la range dans le dictionnaire, puis on la renvoie. Chaque case n'est alors calculée qu'une seule fois.

````python
# Version récursive avec mémoïsation (question 10)
def score_max_memo(i, j, p, memo):
    if (i, j) in memo:
        return memo[(i, j)]
    if i == len(p) - 1:
        res = p[i][j]
    else:
        res = p[i][j] + max(score_max_memo(i + 1, j, p, memo), score_max_memo(i + 1, j + 1, p, memo))
    memo[(i, j)] = res
    return res


ex1 = [[4], [6, 2], [3, 5, 7], [5, 1, 6, 2], [4, 7, 3, 5, 2]]
print(score_max_memo(0, 0, ex1, {}))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La mémoïsation, c'est « ne jamais refaire deux fois le même calcul ». On l'a déjà vue avec Fibonacci (exercice 11).
