### Fiche R5 — Écrire une fonction récursive

🔎 **Je la reconnais :** « Écrire une fonction récursive… », « En utilisant les questions précédentes, écrire une fonction récursive… », ou bien l'énoncé donne une **relation** : `N(m, n) = N(m - 1, n) + N(m, n - 1)`, `d(n) = 2 × d(n - 1) + 1`, `score_max(i, j, p) = p[i][j] + max(...)`.

🛠️ **La méthode des 3 questions** (à écrire au brouillon, en français, **avant** le code) :
1. **Quel est le cas le plus petit, dont je connais la réponse tout de suite ?** → c'est le **cas de base**.
2. **Comment je me ramène à un problème plus petit ?** → `n - 1`, `n // 2`, l'indice suivant `i + 1`, la liste sans son premier élément… → c'est **l'appel récursif**.
3. **Si quelqu'un me donne la réponse du petit problème, comment je fabrique la réponse du grand ?** → c'est le `return`.

Puis on recopie le **squelette** :
```python
def f(parametres):
    if cas_de_base:
        return reponse_directe
    else:
        return ... f(parametres_plus_petits) ...
```
Si l'énoncé donne une formule, **le travail est déjà fait** : la formule est le `return`, et les valeurs données « à part » sont les cas de base.

📌 **Exemple.** Écrire une fonction récursive `puissance(x, n)` qui renvoie x à la puissance n (n entier positif).

| Question | Réponse au brouillon |
|:---|:---|
| 1. Le plus petit cas ? | n = 0 : x⁰ = 1 |
| 2. Plus petit ? | `puissance(x, n - 1)` |
| 3. Comment je fabrique la réponse ? | xⁿ = x × xⁿ⁻¹, donc `x * puissance(x, n - 1)` |

````python
def puissance(x, n):
    if n == 0:
        return 1
    else:
        return x * puissance(x, n - 1)


print(puissance(2, 10))      # 1024
````

⚠️ **Pièges relevés dans la base :**
- oublier le `return` devant l'appel récursif : la fonction calcule… puis renvoie `None` ;
- un cas de base qu'on ne rencontre jamais (avec `n - 2`, un `n` impair saute par-dessus `0` : exercice 4) ;
- oublier un cas de base quand il y a **deux** appels (Fibonacci recule de 1 **et** de 2 : il faut deux cas de base) ;
- avec une liste, utiliser un **indice** qui avance (`i + 1`) est plus simple que de fabriquer des sous-listes ; le cas de base est alors « l'indice est à la fin ».

### Fiche R2 — Prévoir ce qu'affiche ou renvoie un appel : l'arbre d'appels

🔎 **Je la reconnais :** « Qu'affiche la commande `f(5)` ? », « Écrire ce qui sera affiché… », « Quelle valeur sera affectée à `res` ? », « Combien de fois est appelée la fonction ? (on comptera l'appel initial) », « Donner les affichages produits par… », « Remplir le tableau ».

🛠️ **La méthode : l'arbre d'appels.** C'est la méthode de la ressource officielle Eduscol « Récursivité » (Terminale NSI, pages 5 et 6) : on dessine l'**arbre d'appels**, puis on le parcourt « **à rebours** ».

**Chaque nœud de l'arbre est un appel de fonction.** Dans chaque nœud, j'écris **les paramètres** et **le retour** : `f(4) → ?`

1. **Je descends.** J'écris l'appel de départ. Sous chaque nœud, j'écris les appels qu'il fait, dans l'ordre du code (un appel placé dans une boucle donne un enfant par tour de boucle). Je **laisse le retour vide** (`?`) : je ne le connais pas encore.
2. **Je numérote** les nœuds dans l'ordre où les appels sont faits : ①, ②, ③…
3. **J'arrive au cas de base** : là, le retour est connu tout de suite. Je l'écris.
4. **Je remonte.** Dès que tous les enfants d'un nœud ont leur retour, je calcule le retour de ce nœud avec sa ligne `return`, et je l'écris à la place du `?`. Je continue jusqu'à l'appel de départ.

**Les affichages (`print`)** : je les note à côté du nœud, au moment où ils se font :
- un `print` placé **avant** l'appel récursif s'écrit pendant la **descente** ;
- un `print` placé **après** l'appel récursif s'écrit pendant la **remontée**.

**« Combien d'appels ? »** : c'est le nombre de nœuds, donc le dernier numéro (appel initial compris ; si l'énoncé dit « sans compter l'appel initial », on enlève 1).

**S'il y a plusieurs appels** (Fibonacci, tri fusion, baguenaudier, un appel dans une boucle) : un nœud a plusieurs enfants. Le premier enfant est **entièrement fini** (descente et remontée) avant que le deuxième commence : c'est l'ordre des numéros.

📌 **Exemple.** Que renvoie `f(4)`, et qu'affiche-t-il ?

````python
def f(n):
    if n == 0:
        return 0
    print(n)
    return n + f(n - 2)


res = f(4)
print("res =", res)
````

**Au brouillon, la descente** (les retours sont encore vides) :
```
① f(4) → ?             affiche 4
└── ② f(2) → ?         affiche 2
    └── ③ f(0) → 0     cas de base : le retour est connu
```
**Puis la remontée** (on remplit les retours de bas en haut) :
```
① f(4) → 4 + 2 = 6
└── ② f(2) → 2 + 0 = 2
    └── ③ f(0) → 0
```
Affichage : `4` puis `2` (le `print` est avant l'appel : il se fait pendant la descente). Valeur : `res = 6`. Nombre d'appels : 3.

⚠️ **Pièges relevés dans la base :**
- un `print` placé **avant** l'appel récursif affiche en descendant (4, 2) ; placé **après**, il affiche en remontant (2, 4) ;
- « combien d'appels » : on compte **tous les nœuds**, **appel initial compris** (le sujet le rappelle souvent) ;
- avec plusieurs appels, ne pas lire l'arbre « étage par étage » : on suit les **numéros** ;
- remplir un tableau de valeurs (`f(1)`, `f(2)`… `f(12)`) : on **réutilise** les cases déjà calculées, on ne redessine pas tout l'arbre ;
- si l'énoncé fournit **son propre tableau** (ou son arbre) à compléter, on répond dans celui-ci : notre arbre reste au brouillon.

### Pour suivre une boucle : le tableau de trace

L'arbre d'appels sert pour les **fonctions récursives**. Quand le code contient une **boucle** (`for`, `while`), on suit les variables avec un **tableau de trace** : une colonne par variable, une ligne par tour de boucle.

| Étape | variable 1 | variable 2 | … | condition du `while` |
|:---|:---|:---|:---|:---|
| avant la boucle | … | … | | … |
| fin du tour 1 | … | … | | … |

- La première ligne : les valeurs **avant** la boucle.
- Puis une ligne **à la fin de chaque tour**, avec les nouvelles valeurs.
- Pour un `while`, une colonne avec la condition : on s'arrête à la première ligne où elle vaut `False`.
- Pour un `for`, j'écris d'abord **la liste des valeurs** que va prendre la variable de boucle (⚠️ `range(a, b)` s'arrête **avant** `b`) : il y aura une ligne par valeur.

Les exemples ci-dessous couvrent les boucles qu'on rencontre dans les sujets : `while`, `for` sur un `range`, `for` sur les valeurs d'un tableau, `for` sur les indices d'un tableau, et deux boucles l'une dans l'autre.

📌 **Exemple 1 — une boucle `while`.** Que renvoie `g(3)` ?

````python
def g(n):
    s = 0
    i = 1
    while i <= n:
        s = s + i
        i = i + 1
    return s


print(g(3))
````

| Étape | `s` | `i` | `i <= n` |
|:---|:---:|:---:|:---:|
| avant la boucle | 0 | 1 | `True` |
| fin du tour 1 | 1 | 2 | `True` |
| fin du tour 2 | 3 | 3 | `True` |
| fin du tour 3 | 6 | 4 | `False` : on sort |

`g(3)` renvoie **6**.

📌 **Exemple 2 — une boucle `for` sur un `range`.** Que renvoie `factorielle(4)` ?

````python
def factorielle(n):
    p = 1
    for i in range(1, n + 1):
        p = p * i
    return p


print(factorielle(4))
````

Valeurs prises par `i` : `range(1, 5)` donne **1, 2, 3, 4** (5 n'est pas pris).

| Étape | `i` | `p` |
|:---|:---:|:---:|
| avant la boucle | — | 1 |
| fin du tour 1 | 1 | 1 × 1 = 1 |
| fin du tour 2 | 2 | 1 × 2 = 2 |
| fin du tour 3 | 3 | 2 × 3 = 6 |
| fin du tour 4 | 4 | 6 × 4 = 24 |

`factorielle(4)` renvoie **24**.

📌 **Exemple 3 — une boucle `for` qui parcourt les valeurs d'un tableau.** Que renvoie `maximum([3, 7, 2, 9])` ?

````python
def maximum(tab):
    m = tab[0]
    for x in tab:
        if x > m:
            m = x
    return m


print(maximum([3, 7, 2, 9]))
````

Valeurs prises par `x` : les **éléments** du tableau, dans l'ordre : **3, 7, 2, 9**.

| Étape | `x` | `x > m` ? | `m` |
|:---|:---:|:---:|:---:|
| avant la boucle | — | | 3 |
| fin du tour 1 | 3 | 3 > 3 ? non | 3 |
| fin du tour 2 | 7 | 7 > 3 ? oui | 7 |
| fin du tour 3 | 2 | 2 > 7 ? non | 7 |
| fin du tour 4 | 9 | 9 > 7 ? oui | 9 |

`maximum([3, 7, 2, 9])` renvoie **9**.

Avec `for x in tab`, `x` prend directement les **valeurs** : il n'y a pas d'indice. C'est pareil pour une chaîne : dans `for c in "kayak"`, `c` prend chaque lettre.

📌 **Exemple 4 — une boucle `for` qui parcourt les indices d'un tableau (`range(len(...))`).** Que renvoie `indice_maximum([3, 7, 2, 9])` ?

````python
def indice_maximum(tab):
    im = 0
    for i in range(len(tab)):
        if tab[i] > tab[im]:
            im = i
    return im


print(indice_maximum([3, 7, 2, 9]))
````

Valeurs prises par `i` : `len(tab)` vaut 4, et `range(4)` donne les **indices** **0, 1, 2, 3**. On ajoute une colonne `tab[i]` pour lire la valeur à cet indice.

| Étape | `i` | `tab[i]` | `tab[i] > tab[im]` ? | `im` |
|:---|:---:|:---:|:---:|:---:|
| avant la boucle | — | — | | 0 |
| fin du tour 1 | 0 | 3 | 3 > 3 ? non | 0 |
| fin du tour 2 | 1 | 7 | 7 > 3 ? oui | 1 |
| fin du tour 3 | 2 | 2 | 2 > 7 ? non | 1 |
| fin du tour 4 | 3 | 9 | 9 > 7 ? oui | 3 |

`indice_maximum([3, 7, 2, 9])` renvoie **3** (la position du 9, pas sa valeur).

**Exemple 3 ou exemple 4 ?** Avec `for x in tab`, on a les **valeurs**. Avec `for i in range(len(tab))`, on a les **indices** : il faut cette forme quand on veut une **position**, quand on **modifie** `tab[i]`, ou quand on compare deux voisins `tab[i]` et `tab[i + 1]`.

📌 **Exemple 5 — deux boucles l'une dans l'autre.** Qu'affiche ce programme ?

````python
for i in range(1, 3):
    for j in range(1, 4):
        print(i * j)
````

Valeurs prises : `i` dans **1, 2** ; pour **chaque** valeur de `i`, `j` dans **1, 2, 3**. La boucle intérieure fait **tous** ses tours avant que `i` change : une ligne par tour de la boucle intérieure.

| Étape | `i` | `j` | affiche `i * j` |
|:---|:---:|:---:|:---:|
| tour 1 | 1 | 1 | 1 |
| tour 2 | 1 | 2 | 2 |
| tour 3 | 1 | 3 | 3 |
| tour 4 | 2 | 1 | 2 |
| tour 5 | 2 | 2 | 4 |
| tour 6 | 2 | 3 | 6 |

Le programme affiche `1`, `2`, `3`, `2`, `4`, `6` (une valeur par ligne) : 2 × 3 = **6** affichages.

### Fiche R3 — Terminaison : pourquoi ça s'arrête (ou pas)

🔎 **Je la reconnais :** « Expliquer pourquoi le programme s'arrête… », « Justifier que l'appel se terminera quelle que soit… », « L'appel `f2(5)` termine-t-il ? », « Expliquer pourquoi ce cas amènerait une erreur », « Ce code est-il adapté à 2000 cases ? », « Que va-t-il se passer avec n = 10000 ? ».

🛠️ **La méthode :**
1. Repérer la **quantité qui diminue** à chaque appel : `n`, la longueur de la chaîne, `j - i`, le nombre de cases…
2. Vérifier qu'en diminuant, elle **tombe forcément** sur le cas de base (attention aux pas de 2, aux nombres négatifs).
3. Pour montrer que ça **ne** termine **pas**, donner la suite des appels : `f2(5)` → `f2(3)` → `f2(1)` → `f2(-1)` → … on ne passe jamais par 0.

**Phrase type :** « La fonction se termine car à chaque appel, `n` diminue de 1 ; il finit donc par valoir 0, qui est le cas de base, et là la fonction ne s'appelle plus. »

**`RecursionError`** : Python garde en mémoire les appels **en attente** ; au-delà d'environ **1000**, il s'arrête avec `RecursionError: maximum recursion depth exceeded`. Une fonction peut donc être juste en théorie, mais inutilisable pour 2000 ou 10000 (baguenaudier, Fibonacci).

⚠️ **Pièges relevés dans la base :** une boucle `while i != 10` à partir de 12 ne s'arrête jamais (exercice 4, partie A) ; un tirage au hasard (`choice`) qui peut « reculer l'arrêt » indéfiniment (exercice 14).

### Fiche R4 — Compléter une fonction à trous

🔎 **Je la reconnais :** « Recopier et compléter… », « Compléter les lignes 4, 11 et 13 », « Choisir parmi les propositions celle qui convient », du code avec des `...` ou des `à compléter`.

🛠️ **La méthode :**

### Fiche R6 — Écrire la version itérative (avec une boucle)

🔎 **Je la reconnais :** « Écrire une fonction non récursive… », « Écrire une version itérative… », « Proposer une version itérative… ».

🛠️ **La méthode : la récursive donne déjà tout.**

| Dans la version récursive | Dans la version avec boucle |
|:---|:---|
| la valeur du cas de base | la **valeur de départ** de l'accumulateur |
| l'opération du `return` (`n + ...`, `x * ...`) | ce qu'on fait **à chaque tour** de boucle |
| les appels f(n), f(n − 1), …, f(0) | les tours de boucle (`for k in range(...)`) |

````python
def somme_rec(n):                # récursive
    if n == 0:
        return 0
    return n + somme_rec(n - 1)


def somme_boucle(n):             # itérative : même départ (0), même opération (+)
    total = 0
    for k in range(1, n + 1):
        total = total + k
    return total


print(somme_rec(100), somme_boucle(100))
````

⚠️ **Pièges :** pour un palindrome, on compare `chaine[i]` et `chaine[len(chaine) - 1 - i]` jusqu'au milieu ; pour Fibonacci, il faut **deux** variables qui gardent les deux derniers termes.

### Fiche R7 — Coût, nombre d'appels, calculs en double

🔎 **Je la reconnais :** « Expliquer pourquoi le nombre d'appels augmente aussi vite », « Justifier que tester tous les conduits n'est pas raisonnable », « Quel est le nombre d'appels récursifs ? », « Expliquer comment éviter la redondance des calculs ».

🛠️ **La méthode :** dessiner le début de l'**arbre d'appels** et repérer les appels **identiques** (pour `fibonacci(5)`, `fibonacci(3)` est calculé deux fois, `fibonacci(2)` trois fois). Quand chaque appel en fait deux, le nombre d'appels **double à chaque niveau** : de l'ordre de 2ⁿ.

**Les remèdes à citer :** une version avec une boucle, ou garder les résultats déjà calculés dans un tableau ou un dictionnaire (**mémoïsation**, c'est la programmation dynamique).

````python
nb_appels = 0

def fib(n):
    global nb_appels
    nb_appels = nb_appels + 1
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)


for n in [10, 20, 25]:
    nb_appels = 0
    fib(n)
    print("fib(" + str(n) + ") :", nb_appels, "appels")
````

## 5. Les 4 formes de fonctions récursives qu'on rencontre

Toutes les fonctions de la base entrent dans l'une de ces 4 formes. Reconnaître la forme, c'est avoir déjà la moitié du code.

| Forme | Le paramètre qui diminue | Squelette | Dans la base |
|:---|:---|:---|:---|
| **1. Sur un entier** | `n` → `n - 1` ou `n // 2` | `if n == 0: return ...` / `return ... f(n - 1)` | factorielle, somme, compte à rebours, binaire, Syracuse, multiplication |
| **2. Sur une liste ou une chaîne** | un indice `i` → `i + 1` (ou la chaîne raccourcie) | `if i == len(t): return ...` / `return ... t[i] ... f(t, i + 1)` | palindrome, chiffres romains, rendu de monnaie, règles A et B |
| **3. Plusieurs appels** | plusieurs paramètres, plusieurs appels | `return f(...) + f(...)` | Fibonacci, chemins dans une grille, baguenaudier, zone d'une image |
| **4. Un choix à faire** | l'indice de l'objet qu'on envisage | `option1 = f(sans)` / `option2 = f(avec)` / garder la meilleure | charger un camion, forer un puits |

*La forme « couper en deux moitiés » est le chapitre suivant : diviser pour régner.*
