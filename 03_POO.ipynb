⚠️ Dans un sujet, **les questions s'enchaînent** : une fonction écrite à la question 3 sert souvent à la question 5. Si la question dit « en utilisant… », on **appelle** la fonction précédente, on ne la réécrit pas.

### Fiche P5 — Écrire une méthode qui lit ou modifie les attributs

🔎 **Je la reconnais :** « Écrire la méthode `crediter` », « Écrire une méthode `raz` permettant de mettre toutes les coordonnées à zéro », « La méthode `get_annee` renvoie la valeur de l'attribut `annee` », « Écrire la méthode `est_positif` ».

🛠️ **Les 4 sortes de méthodes simples** (repérer laquelle on demande) :

| Sorte | Elle… | Elle se termine par… | Exemple |
|:---|:---|:---|:---|
| **get** (lire) | renvoie un attribut | `return self.attribut` | `get_note` |
| **set** (modifier) | change un attribut avec le paramètre | rien (pas de `return`) | `set_mois`, `changer_role` |
| **test** | répond vrai ou faux | `return` une comparaison | `est_positif`, `est_dispo` |
| **calcul / action** | calcule à partir des attributs, ou les modifie | `return` le résultat, ou rien | `estim_prix`, `crediter`, `avancex` |

**Le squelette :**
```python
    def nom_methode(self, parametres):
        ... self.attribut ...          # on travaille TOUJOURS avec self.
        return ...                     # seulement si la méthode doit renvoyer quelque chose
```

````python
class Compte:
    def __init__(self, titulaire):
        self.titulaire = titulaire
        self.solde = 0

    def get_solde(self):               # get
        return self.solde

    def crediter(self, montant):       # action : modifie, ne renvoie rien
        self.solde = self.solde + montant

    def est_positif(self):             # test
        return self.solde >= 0


c = Compte("Martin")
c.crediter(200)
print(c.get_solde(), c.est_positif())
````

⚠️ **Pièges relevés dans la base :**
- oublier `self.` : `solde = solde + montant` crée une variable locale et ne modifie pas l'objet ;
- oublier `self` comme **premier paramètre** dans le `def` ;
- mettre un `return` dans une méthode qui doit **modifier** (ou l'inverse : modifier un attribut alors qu'on demande de **renvoyer** une valeur) ;
- une méthode « avec limite » (20 points maximum, solde suffisant) : on **teste avant** de modifier.

🔎 **Je la reconnais :** « Écrire l'instruction permettant de créer l'objet… », « Parmi les instructions suivantes, laquelle permet de déclarer… », « Écrire un programme permettant : a. la création d'une instance… c. à lancelot d'attaquer le sorcier… », « Écrire la ligne de code permettant de déposer 200 € ».

🛠️ **Les trois écritures à connaître :**

| Je veux… | J'écris | Exemple |
|:---|:---|:---|

Les arguments sont dans **l'ordre des paramètres du constructeur** (on saute `self`).

⚠️ **Pièges des QCM de la base :** `Carte(self, 7, "coeur")` (le `self` ne s'écrit jamais à l'appel) ; `new Joueur[...]` (c'est d'un autre langage) ; `Joueur{...}` ou une liste `[...]` (ce n'est pas un objet).

### Fiche P6 — Une méthode qui utilise une autre méthode, ou un autre objet

🔎 **Je la reconnais :** « On utilisera les autres méthodes de la classe sans en modifier directement les attributs », « en veillant à ne pas accéder directement aux attributs d'une maison depuis la classe `Antenne` », « prend en paramètre une carte `autre` (de type `Carte`) », « en utilisant les méthodes des classes `Joueur` et `Base` ».

🛠️ **La méthode :**
- appeler une méthode **du même objet** : `self.autre_methode(...)` ;
- utiliser un **autre objet** reçu en paramètre : `autre.attribut` ou, mieux, `autre.methode(...)` ;
- si l'énoncé dit « sans accéder directement aux attributs », on passe **par les méthodes** (`get_...`).

````python
def transferer(self, autre_compte, montant):
    self.solde = self.solde - montant           # interdit si l'énoncé dit « sans modifier les attributs »
    autre_compte.crediter(montant)


def transferer_bien(self, autre_compte, montant):
    self.crediter(-montant)                      # on passe par les méthodes
    autre_compte.crediter(montant)


Compte.transferer = transferer_bien
a, b = Compte("A"), Compte("B")
a.crediter(100)
a.transferer(b, 30)
print(a.get_solde(), b.get_solde())
````

⚠️ **Piège :** dans une méthode qui reçoit `autre`, bien distinguer `self.valeur` (moi) et `autre.valeur` (l'autre). Faire un petit dessin avec deux objets aide beaucoup.

### Fiche P7 — Une fonction qui parcourt une liste d'objets

🔎 **Je la reconnais :** « Écrire une fonction qui prend en paramètre une liste d'objets de la classe `Colis` et qui renvoie… le nombre de… / le poids total… / une nouvelle liste contenant… / le concurrent le plus rapide… / le titre si l'identifiant existe, `None` sinon ».

🛠️ **On reconnaît l'un des 6 modèles**, et on écrit `objet.attribut` dans la boucle :

| Modèle | La question dit… | Squelette |
|:---|:---|:---|
| **compter** | « le nombre de… » | `n = 0` / `for o in liste: if condition: n = n + 1` |
| **additionner** | « le total », « le cumul » | `total = 0` / `total = total + o.attribut` |
| **filtrer** | « une nouvelle liste contenant les… » | `res = []` / `if condition: res.append(o)` |
| **chercher** | « renvoie … si … existe, `None` sinon » | `for o in liste: if condition: return ...` / après la boucle : `return None` |
| **le meilleur** | « le plus rapide », « le plus endetté » | `meilleur = liste[0]` / `if o.attr < meilleur.attr: meilleur = o` |
| **construire** | « renvoie une liste de 52 objets », « crée la liste des maisons » | `res = []` / `res.append(Classe(...))` |

````python
class Colis:
    def __init__(self, ident, poids, etat):
        self.ident = ident
        self.poids = poids
        self.etat = etat


stock = [Colis(1, 12, 'transit'), Colis(2, 30, 'livré'), Colis(3, 5, 'transit')]

total = 0                                   # additionner
for c in stock:
    total = total + c.poids

en_transit = []                             # filtrer
for c in stock:
    if c.etat == 'transit':
        en_transit.append(c)

plus_lourd = stock[0]                       # le meilleur
for c in stock:
    if c.poids > plus_lourd.poids:
        plus_lourd = c

print(total, len(en_transit), plus_lourd.ident)
````

⚠️ **Pièges :** dans le modèle « chercher », le `return None` va **après** la boucle (pas dans un `else`) ; dans « le meilleur », on garde l'**objet** (`meilleur = c`) pour pouvoir renvoyer ce qu'on demande (l'objet, son nom…).

## Exercice 1 — La bataille  ⭐

📚 **Origine :** épreuve écrite, **Centres étrangers 2022, jour 2**, exercice 4

*Cet exercice porte sur les structures de données (programmation objet).*

Simon souhaite créer en Python le jeu de cartes « la bataille » pour deux joueurs. Les questions qui suivent demandent de reprogrammer quelques fonctions du jeu.

On rappelle ici les règles du jeu de la bataille :

**Préparation**
- Distribuer toutes les cartes aux deux joueurs.
- Les joueurs ne prennent pas connaissance de leurs cartes et les laissent en tas face cachée devant eux.

**Déroulement**
- A chaque tour, chaque joueur dévoile la carte du haut de son tas.
- Le joueur qui présente la carte ayant la plus haute valeur emporte les deux cartes qu'il place sous son tas.
- Les valeurs des cartes sont : dans l'ordre de la plus forte à la plus faible : As, Roi, Dame, Valet, 10, 9, 8, 7, 6, 5, 4, 3 et 2 (la plus faible)

Si deux cartes sont de même valeur, il y a “bataille”.

## Exercice 2 — Le code de César  ⭐

📚 **Tombé au bac :** épreuve écrite, **Centres étrangers 2021, sujet 1**, exercice 1 (repris aussi en contrôle par des professeurs)

*Notion abordée : programmation objet.*

Dans cet exercice, on étudie une méthode de chiffrement de chaînes de caractères alphabétiques. Pour des raisons historiques, cette méthode de chiffrement est appelée « code de César ». On considère que les messages ne contiennent que les lettres capitales de l'alphabet `"ABCDEFGHIJKLMNOPQRSTUVWXYZ"` et la méthode de chiffrement utilise un nombre entier fixé appelé la clé de chiffrement.

*La classe `CodeCesar` donnée par le sujet (à exécuter) :*

````python
class CodeCesar:

    def __init__(self, cle):
        self.cle = cle
        self.alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

    def decale(self, lettre):
        num1 = self.alphabet.find(lettre)
        num2 = num1 + self.cle
        if num2 >= 26:
            num2 = num2 - 26
        if num2 < 0:
            num2 = num2 + 26
        nouvelle_lettre = self.alphabet[num2]
        return nouvelle_lettre
````

🏷️ *Type de question :* **P4** Prévoir la valeur d'un attribut ou ce que renvoie un code

#### 📝 ÉNONCÉ

> On rappelle que la méthode `str.find(lettre)` renvoie l'indice (index) de la lettre dans la chaîne de caractères `str`. Représenter le résultat d'exécution du code Python suivant :
> ```python
> code1 = CodeCesar(3)
> print(code1.decale('A'))
> print(code1.decale('X'))
> ```

#### ✅ RÉSOLUTION

```
D
A
```

`'A'` a l'indice 0, 0 + 3 = 3 → `'D'`. `'X'` a l'indice 23, 23 + 3 = 26 ⩾ 26 donc 26 − 26 = 0 → `'A'`.

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Prévoir le résultat d'une méthode** : on remplace `self.cle` par la valeur donnée au constructeur (3), puis on suit les lignes une par une (tableau de trace avec `num1`, `num2`).
- Le test `num2 >= 26` fait **reboucler** l'alphabet : après Z, on revient à A.

### Question 2

🏷️ *Type de question :* **P6** Écrire une méthode qui utilise une autre méthode ou un autre objet

#### 📝 ÉNONCÉ

> La méthode de chiffrement du « code César » consiste à décaler les lettres du message dans l'alphabet d'un nombre de rangs fixé par la clé. Par exemple, avec la clé 3, toutes les lettres sont décalées de 3 rangs vers la droite : le A devient le D, le B devient le E, etc.
>
> Ajouter une méthode `cryptage(self, texte)` dans la classe `CodeCesar` définie à la question précédente, qui reçoit en paramètre une chaîne de caractères (le message à crypter) et qui retourne une chaîne de caractères (le message crypté). Cette méthode `cryptage(self, texte)` doit crypter la chaîne `texte` avec la clé de l'objet de la classe `CodeCesar` qui a été instancié.
>
> Exemple :
> ```python
> >>> code1 = CodeCesar(3)
> >>> code1.cryptage("NSI")
> 'QVL'
> ```
> *Dans le notebook, la dernière ligne ajoute la méthode à la classe.*

#### ✅ RÉSOLUTION

````python
def cryptage(self, texte):
    resultat = ''
    for lettre in texte:
        resultat = resultat + self.decale(lettre)
    return resultat


CodeCesar.cryptage = cryptage
````

*Les tests de la feuille d'exercices :*

````python
code1 = CodeCesar(3)
assert code1.cryptage("NSI") == "QVL"
assert code1.cryptage("XYZ") == "ABC"
assert CodeCesar(-3).cryptage("QVL") == "NSI"
assert code1.cryptage("") == ""
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est une méthode **qui utilise une autre méthode** de la même classe : on écrit `self.decale(lettre)`, avec `self.` devant (sans `self.`, Python ne trouve pas `decale`).
- Modèle « accumuler » sur une chaîne : on part de `''` et on ajoute une lettre à chaque tour.

Il s'affiche **`FIN`**. La méthode change le signe de la clé (10 devient −10), crypte, puis remet la clé à 10. Crypter avec −10 décale chaque lettre de 10 rangs **vers la gauche** : P → F, S → I, X → N. `transforme` sert donc à **décrypter** un message crypté avec la clé 10.

````python
def transforme(self, texte):
    self.cle = -self.cle
    message = self.cryptage(texte)
    self.cle = -self.cle
    return message


CodeCesar.transforme = transforme
print(CodeCesar(10).transforme("PSX"))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

````python
class Joueur:
    def __init__(self, pseudo, identifiant, equipe):
        self.pseudo = pseudo
        self.equipe = equipe
        self.id = identifiant
        self.nb_de_tirs_emis = 0
        self.liste_id_tirs_recus = []
        self.est_actif = True

    def tire(self):
        if self.est_actif == True:
            self.nb_de_tirs_emis = self.nb_de_tirs_emis + 1

    def est_determine(self):
        return self.nb_de_tirs_emis > 500

    def subit_un_tir(self, id_recu):
        if self.est_actif == True:
            self.est_actif = False
            self.liste_id_tirs_recus.append(id_recu)

    def redevenir_actif(self):
        if self.est_actif == False:
            self.est_actif = True

    def nb_de_tirs_recus(self):
        return len(self.liste_id_tirs_recus)
````

>
> 1. Écrire la fonction `rechercher_debiteurs` qui prend en argument `liste_comptes` et renvoie une liste de dictionnaires dont la première clé est le nom de l'adhérent et la seconde clé le solde de son compte en négatif.
>
>    Exemple de résultat :
>
>    `liste_debiteurs = [{'Nom': 'DUPONT Thomas', 'solde': -60}, {'Nom': 'CARNEIRO Sarah', 'solde': -75}]`.
> 2. Écrire la fonction `urgent_debiteur` qui prend en argument la liste de dictionnaires et renvoie le nom de l'adhérent le plus endetté. On suppose qu'aucun adhérent n'a la même dette.

#### ✅ RÉSOLUTION

````python
def rechercher_debiteurs(liste_comptes):
    debiteurs = []
    for c in liste_comptes:
        if not c.est_positif():
            debiteurs.append({'Nom': c.adherent, 'solde': c.solde})
    return debiteurs


def urgent_debiteur(liste_debiteurs):
    pire = liste_debiteurs[0]
    for d in liste_debiteurs:
        if d['solde'] < pire['solde']:
            pire = d
    return pire['Nom']
````

````python
import random


class Personnage:
    def __init__(self, coordx, coordy, coordz, point_de_vie, point_degats):
        self.x = coordx
        self.y = coordy
        self.z = coordz
        self.vie = point_de_vie
        self.degats = point_degats

    def avancex(self):
        self.x = self.x + 1

    def raz(self):
        self.x = 0
        self.y = 0
        self.z = 0

    def coord(self):
        return (self.x, self.y, self.z)

    def get_etat(self):
        return self.vie

    def newgame(self):
        if self.vie <= 0:
            self.raz()
            self.vie = 15

    def potionmystere(self):
        if random.randint(1, 2) == 1:
            nbPoint = -1
        else:
            nbPoint = +1
        self.vie = self.vie + nbPoint

    def piege(self):
        self.vie = self.vie - 10

    def repos(self):
        self.vie = self.vie + 5

    def perdre_vie(self, points):
        self.vie = self.vie - points
        self.newgame()

    def attaquer(self, autre):
        autre.perdre_vie(self.degats)
````

````python
class Region:
    def __init__(self, nom_region):
        self.nom = nom_region
        self.tab_voisines = []
        self.tab_couleurs_disponibles = ['rouge', 'vert', 'bleu', 'jaune', 'orange', 'marron']
        self.couleur_attribuee = None

    def renvoie_premiere_couleur_disponible(self):
        return self.tab_couleurs_disponibles[0]

    def renvoie_nb_voisines(self):
        return len(self.tab_voisines)

    def est_coloriee(self):
        return self.couleur_attribuee != None

    def retire_couleur(self, couleur):
        if couleur in self.tab_couleurs_disponibles:
            self.tab_couleurs_disponibles.remove(couleur)

    def est_voisine(self, region):
        for voisine in self.tab_voisines:
            if voisine == region:
                return True
        return False
````

#### 📝 ÉNONCÉ

> Le code Python ci-après présente la fonction récursive `chargement_glouton` dont les paramètres sont :
> - `liste` : une liste de colis triées par poids décroissants ;
> - `rang` : un indice compris entre 0 inclus et `len(liste)` inclus ;
> - `charge` : une charge exprimée en kilogrammes ;
>
> et qui renvoie la liste des colis à charger en appliquant l'algorithme glouton, en supposant que la charge restante dans le camion est capacité, et en ne considérant que les colis de liste d'indice supérieur ou égal à `rang`.
> ```python
> 1  def chargement_glouton(liste, rang, capacite):
> 2      if rang == len(liste):
> 3          return ...
> 4      elif liste[rang].poids <= ...:
> 5          return ... + chargement_glouton(liste, ..., ...)
> 6      else:
> 7          return chargement_glouton(liste, ..., ...)
> ```
> 8. Recopier et compléter le code ci-dessus de la fonction `chargement_glouton`.

#### ✅ RÉSOLUTION

````python
def chargement_glouton(liste, rang, capacite):
    if rang == len(liste):
        return []
    elif liste[rang].poids <= capacite:
        return [liste[rang]] + chargement_glouton(liste, rang + 1, capacite - liste[rang].poids)
    else:
        return chargement_glouton(liste, rang + 1, capacite)
````

  - **le colis tient** : on le prend (`[liste[rang]]`, une liste d'un seul élément) et on continue avec le colis suivant et une **capacité diminuée** de son poids ;
  - **le colis ne tient pas** : on passe au suivant, la capacité ne change pas.
- `[liste[rang]] + ...` colle deux listes : il faut des crochets autour de `liste[rang]`.

> On pourra créer une liste `colis_a_charger`, puis parcourir les colis triés en les ajoutant à cette liste tant que le poids total n'excède pas la capacité du camion.

#### ✅ RÉSOLUTION

````python
def chargement_glouton2(liste, capacite):
    colis_a_charger = []
    for c in liste:
        if c.poids <= capacite:
            colis_a_charger.append(c)
            capacite = capacite - c.poids
    return colis_a_charger
````

🏷️ *Type de question :* **P7** Écrire une fonction qui parcourt une liste d'objets

#### 📝 ÉNONCÉ

> On souhaite créer une fonction `meilleur_concurrent` qui prend en paramètre une liste L de concurrents et qui renvoie l'objet `Concurrent` correspondant au concurrent le plus rapide. On suppose que la liste est non vide.
>
> Recopier et compléter le code Python, donné ci-dessous, de la fonction `meilleur_concurrent`.
> ```python
> 1   def meilleur_concurrent(L):
> 2       conc_mini = L. ...
> 3       mini = conc_mini.temps_tot
> 4       Q = L.queue()
> 5       while not (Q.est_vide()):
> 6           elt = Q.tete()
> 7           if elt.temps_tot ... mini:
> 8               conc_mini = elt
> 9               mini = elt.temps_tot
> 10          Q = Q. ...
> 11      return ...
> ```

#### ✅ RÉSOLUTION

````python
def meilleur_concurrent(L):
    conc_mini = L.tete()
    mini = conc_mini.temps_tot
    Q = L.queue()
    while not (Q.est_vide()):
        elt = Q.tete()
        if elt.temps_tot < mini:
            conc_mini = elt
            mini = elt.temps_tot
        Q = Q.queue()
    return conc_mini
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Modèle 5, **le meilleur** : on part du premier concurrent, et on garde chaque concurrent plus rapide (temps **plus petit**, donc `<`).
- La boucle n'est pas un `for` : la classe `Liste` ne se parcourt qu'avec `tete()` et `queue()`. **Ligne 10** : on avance en remplaçant `Q` par sa queue, sinon la boucle ne s'arrêterait jamais.
- **Ligne 11** : on renvoie l'**objet** (`conc_mini`), pas son temps (`mini`).

🏷️ *Type de question :* **P6** Écrire une méthode qui utilise une autre méthode ou un autre objet

#### 📝 ÉNONCÉ

> **Question 8.** Implémenter la méthode `prete()` de la classe `Bibliotheque`.
> 
> **Question 9.** Implémenter la méthode `recupere()` de la classe `Bibliotheque`.

#### ✅ RÉSOLUTION

````python
def prete(self, livre, usager):
    if not livre.estDispo():
        return False
    if usager.nbLivresEmpruntes() >= Bibliotheque.nb_max:
        return False
    self.ajouteLivre(livre)
    self.ajouteUsager(usager)
    livre.dispo = False
    usager.livres[livre.id] = livre
    return True


def recupere(self, livre, usager):
    if livre.id in usager.livres:
        del usager.livres[livre.id]
        livre.dispo = True


Bibliotheque.prete = prete
Bibliotheque.recupere = recupere
````

> def couvre(self, maison):

````python
class Antenne:
    def __init__(self, position, rayon):
        self._position = position
        self._rayon = rayon

    def get_pos_antenne(self):
        return self._position

    def get_rayon(self):
        return self._rayon

    def couvre(self, maison):
        distance = abs(maison.get_pos_maison() - self.get_pos_antenne())
        return distance <= self.get_rayon()
````

### Partie A — Questions 1, 2 et 3

🏷️ *Type de question :* **P2** Compléter le constructeur `__init__` · **P5** Écrire une méthode qui lit ou modifie les attributs (get, set, test, calcul) · **P6** Écrire une méthode qui utilise une autre méthode ou un autre objet

#### 📝 ÉNONCÉ

> **Partie A : La classe Carte**
>

>
> Le constructeur de cette classe prend en paramètre une valeur (de type `int`) comprise entre 1 et 104 et l'affecte à l'attribut `valeur` (de type `int`).
>
> On ajoute l'attribut `TdB` (de type `int`) qui contient le nombre de “têtes de bœuf” qui sera calculé à l'aide de la méthode de classe `calcul_TdB` (définie à la question suivante) qui n'a pas de paramètre.
>
> 1. Écrire le code du constructeur de la classe `Carte`.
>
> Chaque carte possède une ou plusieurs “têtes de bœuf”. Ce nombre est déterminé de la façon suivante :
> - si la valeur de la carte est divisible par 11 alors la carte reçoit 5 “têtes de bœuf”;
> - si la valeur de la carte se termine par 0 alors la carte reçoit 3 “têtes de bœuf”;
> - si la valeur de la carte se termine par 5 alors la carte reçoit 2 “têtes de bœuf”;
> - si la valeur de la carte ne remplit aucune de ces conditions alors elle ne reçoit qu'1 “tête de bœuf”.
>
> Attention, ces règles se cumulent.
>
> Par exemple, 55 est divisible par 11 et se termine par 5 donc la carte 55 comporte 7 “têtes de bœuf”.
>
> 2. Écrire le code de la méthode `calcul_TdB` de la classe `Carte` afin de calculer le nombre de “têtes de bœuf” pour une carte donnée.
>
> On veut ajouter à la classe `Carte` une méthode `est_superieure_a` qui prend en paramètre une carte `autre` (de type `Carte`) et qui renvoie `True` si la valeur de la carte considérée est strictement supérieure à la valeur de la carte `autre`.
>
> 3. Écrire le code de cette méthode.
>
> La classe `Carte` possède également la méthode `difference` qui prend en paramètre une carte `autre` (de type `Carte`) et qui renvoie la valeur absolue de la différence entre les valeurs des deux cartes considérées.

#### ✅ RÉSOLUTION

````python
class Carte:
    def __init__(self, valeur):
        self.valeur = valeur
        self.TdB = self.calcul_TdB()

    def calcul_TdB(self):
        tdb = 0
        if self.valeur % 11 == 0:
            tdb = tdb + 5
        if self.valeur % 10 == 0:
            tdb = tdb + 3
        if self.valeur % 10 == 5:
            tdb = tdb + 2
        if tdb == 0:
            tdb = 1
        return tdb

    def est_superieure_a(self, autre):
        return self.valeur > autre.valeur
````

> 11. Coder la méthode `colorie(self)` de la classe `Pays` qui choisit une couleur pour chaque région du pays de la façon suivante :
> - On récupère la région non coloriée qui possède le plus de voisines.
> - Tant que cette région existe :
>   - La couleur attribuée à cette région est la première couleur disponible dans son tableau de couleurs disponibles.
>   - Pour chaque région voisine de la région : si la couleur choisie est présente dans le tableau des couleurs disponibles de la région voisine alors on la retire de ce tableau.
>   - On récupère à nouveau la région non coloriée qui possède le plus de voisines.

#### ✅ RÉSOLUTION

````python
def colorie(self):
    region = self.renvoie_max()
    while region != None:
        couleur = region.renvoie_premiere_couleur_disponible()
        region.couleur_attribuee = couleur
        for voisine in region.tab_voisines:
            voisine.retire_couleur(couleur)
        region = self.renvoie_max()


Pays.colorie = colorie   # ajoute la méthode à la classe (ligne à garder)
````
